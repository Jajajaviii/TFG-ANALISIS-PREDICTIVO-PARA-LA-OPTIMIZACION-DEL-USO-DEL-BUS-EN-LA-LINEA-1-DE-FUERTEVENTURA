```python
"""
TFG - Script unificado desde TFG.ipynb.
Incluye todas las celdas de codigo en un flujo unico y ejecutable.
"""

# Librerias e imports consolidados
import warnings
import matplotlib
from matplotlib.colors import LinearSegmentedColormap
import os
from matplotlib.patches import Patch
import random
import joblib
from pathlib import Path
from sklearn.base import clone
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.metrics import mean_absolute_error, mean_squared_error
from sklearn.linear_model import Ridge
from sklearn.model_selection import GridSearchCV, TimeSeriesSplit, cross_val_score
from sklearn.metrics import r2_score
from xgboost import XGBRegressor
from sklearn.neural_network import MLPRegressor
import importlib
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import matplotlib.dates as mdates

# Configuracion base
warnings.filterwarnings('ignore')



matplotlib.rcParams.update({
    'figure.dpi': 120,
    'axes.spines.top': False,
    'axes.spines.right': False,
    'axes.titlesize': 13,
    'axes.labelsize': 11,
    'font.family': 'DejaVu Sans',
})


# DATASET_PATH: ruta al CSV del dataset principal.
# Por defecto se asume que el CSV esta en el mismo directorio
# que este script. Modificar si se ubica en otra carpeta.
# En notebooks no existe __file__, así que usamos el directorio actual del kernel\n",
BASE_DIR = Path.cwd().resolve()
DATASET_PATH = BASE_DIR / "TFG_L1_dataset_definitivo.csv.gz"

# Verificacion de existencia antes de cargar
if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"No se encuentra el dataset en: {DATASET_PATH}\n"
        "Asegurate de que 'TFG_L1_dataset_definitivo.csv' esta en el mismo "
        "directorio que este script."
    )

print(f"Dataset encontrado en: {DATASET_PATH}")
print("Libraries loaded OK")


df = pd.read_csv(str(DATASET_PATH), parse_dates=['fecha'])
print(f"Shape: {df.shape[0]:,} filas x {df.shape[1]} columnas")
print(f"Memoria estimada: {df.memory_usage(deep=True).sum()/1e6:.1f} MB")


print("=" * 70)
print("TIPOS DE DATOS")
print("=" * 70)
print(df.dtypes.to_string())


print("=" * 70)
print("ESTADISTICAS BASICAS (columnas numericas)")
print("=" * 70)
df.describe().T.style.format("{:.3f}").background_gradient(cmap='Blues', axis=1)


print("=" * 70)
print("VALORES NULOS")
print("=" * 70)
nulls = df.isnull().sum()
nulls_pct = (nulls / len(df) * 100).round(2)
null_report = pd.DataFrame({'nulos': nulls, 'pct_total': nulls_pct}).query('nulos > 0').sort_values('pct_total', ascending=False)
print(null_report.to_string())
print(f"\nTotal columnas con nulos: {len(null_report)}")

# Visualizacion de nulos
if len(null_report) > 0:
    fig, ax = plt.subplots(figsize=(10, max(3, len(null_report)*0.4)))
    bars = ax.barh(null_report.index, null_report['pct_total'], color='#e63946', edgecolor='white')
    ax.bar_label(bars, fmt='%.1f%%', padding=3, fontsize=9)
    ax.set_title('Porcentaje de valores nulos por columna')
    ax.set_xlabel('% de nulos')
    plt.tight_layout(); plt.show()


print("=" * 70)
print("CARDINALIDAD VARIABLES CATEGORICAS Y BOOLEANAS")
print("=" * 70)
cat_cols  = df.select_dtypes(include='object').columns.tolist()
bool_cols = df.select_dtypes(include='bool').columns.tolist()

for c in cat_cols + bool_cols:
    vals = df[c].value_counts()
    top  = vals.head(5).to_dict()
    print(f"  {c:35s}: {df[c].nunique():4d} unicos | top-5: {top}")


print("=" * 70)
print("PERIODO TEMPORAL")
print("=" * 70)
print(f"  Fecha minima : {df['fecha'].min().date()}")
print(f"  Fecha maxima : {df['fecha'].max().date()}")
print(f"  Dias unicos  : {df['fecha'].nunique()}")
n_expected = (df['fecha'].max() - df['fecha'].min()).days + 1
n_missing  = n_expected - df['fecha'].nunique()
print(f"  Dias esperados (rango continuo): {n_expected}")
print(f"  Dias faltantes en la serie: {n_missing}")


# Granularidad base: fecha x cycle_id x segment_id (cycle_id = trip_id original)
df["cycle_id"] = df["trip_id"].astype(str)
df["is_service_segment"] = df["direction"].isin(["ida", "vuelta"])
df["service_id"] = np.where(
    df["is_service_segment"],
    df["cycle_id"] + "_" + df["direction"].astype(str),
    np.nan,
)

gran = df.groupby(["fecha", "cycle_id", "segment_id"]).size()
duplicates = (gran > 1).sum()
print(f"Registros duplicados (fecha x cycle_id x segment_id): {duplicates}")
print(f"Tuplas unicas: {len(gran):,}")

# Segmentos por ciclo
segs_per_cycle = df.groupby(["fecha", "cycle_id"])["segment_id"].nunique()
print(f"\nSegmentos por ciclo -- min:{segs_per_cycle.min()}  max:{segs_per_cycle.max()}  median:{segs_per_cycle.median()}")
print(segs_per_cycle.value_counts().sort_index().rename("N ciclos").to_frame().to_string())

# Segmentos por expedicion unidireccional (service_id)
segs_per_service = (
    df[df["is_service_segment"]]
    .groupby(["fecha", "service_id"])["segment_id"].nunique()
)
print(f"\nSegmentos por expedicion (service_id) -- min:{segs_per_service.min()}  max:{segs_per_service.max()}  median:{segs_per_service.median()}")
print(segs_per_service.value_counts().sort_index().rename("N expediciones").to_frame().to_string())

# Conteos diarios por unidad operativa
cycles_per_day = df.groupby("fecha")["cycle_id"].nunique()
services_per_day = df[df["is_service_segment"]].groupby("fecha")["service_id"].nunique()

print(f"\nCiclos por dia -- min:{cycles_per_day.min()}  max:{cycles_per_day.max()}  median:{cycles_per_day.median()}")
print(cycles_per_day.describe().round(2).to_string())
print(f"\nExpediciones unidireccionales por dia -- min:{services_per_day.min()}  max:{services_per_day.max()}  median:{services_per_day.median()}")
print(services_per_day.describe().round(2).to_string())

# Validacion estructural esperada del dataset
ratio = (services_per_day / cycles_per_day).dropna()
assert (ratio == 2).all(), "Cada cycle_id debe generar exactamente 2 expediciones operativas (ida/vuelta)."
print("\n[OK] Unidad operativa definida: cycle_id (ciclo) y service_id (expedicion unidireccional).")


# Distancia/tiempo por ciclo completo (cycle_id)
df["total_time_min"] = df["drive_time_min"] + df["dwell_time_next_stop_min"]

dist_per_cycle = df.groupby(["fecha", "cycle_id"])["segment_distance_km"].sum()
time_per_cycle = df.groupby(["fecha", "cycle_id"])["total_time_min"].sum()

# Distancia/tiempo por expedicion unidireccional (service_id, excluye cierre)
service_df = df[df["is_service_segment"]].copy()
dist_per_service = service_df.groupby(["fecha", "service_id"])["segment_distance_km"].sum()
time_per_service = service_df.groupby(["fecha", "service_id"])["total_time_min"].sum()

print("Distancia acumulada por ciclo (km):")
print(dist_per_cycle.describe().round(3).to_string())
print("\nDistancia acumulada por expedicion unidireccional (km):")
print(dist_per_service.describe().round(3).to_string())

fig, axes = plt.subplots(1, 2, figsize=(14, 4))
axes[0].hist(dist_per_cycle, bins=40, color="steelblue", edgecolor="white", alpha=0.9, label="Ciclo")
axes[0].hist(dist_per_service, bins=40, color="orange", edgecolor="white", alpha=0.55, label="Expedicion")
axes[0].set_xlabel("Km"); axes[0].set_ylabel("Frecuencia")
axes[0].set_title("Distribucion distancia: ciclo vs expedicion")
axes[0].legend()

print("\nTiempo total por ciclo (min):")
print(time_per_cycle.describe().round(2).to_string())
print("\nTiempo total por expedicion unidireccional (min):")
print(time_per_service.describe().round(2).to_string())

axes[1].hist(time_per_cycle, bins=40, color="teal", edgecolor="white", alpha=0.9, label="Ciclo")
axes[1].hist(time_per_service, bins=40, color="#ff9f1c", edgecolor="white", alpha=0.55, label="Expedicion")
axes[1].set_xlabel("Minutos"); axes[1].set_ylabel("Frecuencia")
axes[1].set_title("Distribucion tiempo: ciclo vs expedicion")
axes[1].legend()

plt.tight_layout(); plt.show()

print("\n[OK] La unidad operativa para simulacion/KPI se fija en service_id (expedicion).")


# DataFrame a nivel dia
day_uniq = df.drop_duplicates(subset='fecha').set_index('fecha').sort_index()
print(f"Dias en dataset: {len(day_uniq)}")
print("\nDistribucion day_type:")
print(df.drop_duplicates('fecha')['day_type'].value_counts().to_string())


# Distribucion dias por tipo
counts = df.drop_duplicates('fecha')['day_type'].value_counts()
color_map = {
    'laborable':          '#3a86ff',
    'sabado':             '#ff9f1c',
    'domingo':            '#e63946',
    'festivo':            '#8ecae6',
    'festivo_laborable':  '#a8dadc',
}

fig, ax = plt.subplots(figsize=(9, 4))
bars = ax.bar(counts.index, counts.values,
              color=[color_map.get(x, '#888') for x in counts.index], edgecolor='white')
ax.bar_label(bars, padding=3, fontsize=10)
ax.set_title('Numero de dias por tipo de dia')
ax.set_xlabel('Tipo de dia'); ax.set_ylabel('Numero de dias')
plt.tight_layout(); plt.show()


pax_daily  = df.drop_duplicates('fecha').set_index('fecha')['airport_pax'].sort_index()
monthly    = pax_daily.resample('ME').agg(['mean', 'min', 'max'])

fig, ax = plt.subplots(figsize=(14, 5))
ax.fill_between(monthly.index, monthly['min'], monthly['max'],
                alpha=0.2, color='steelblue', label='Rango min/max')
ax.plot(monthly.index, monthly['mean'], 'o-', color='steelblue', lw=2, ms=6, label='Media mensual')
ax.xaxis.set_major_formatter(mdates.DateFormatter('%b\n%Y'))
ax.xaxis.set_major_locator(mdates.MonthLocator())
ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'{x/1e6:.2f}M'))
ax.set_title('Serie temporal mensual de Pasajeros Aeropuerto (airport_pax)')
ax.set_xlabel('Mes'); ax.set_ylabel('Pasajeros aeropuerto'); ax.legend()
plt.tight_layout(); plt.show()

print("\nResumen mensual:")
print(monthly.round(0).to_string())


day_pax = df.drop_duplicates('fecha')[['day_type', 'airport_pax']].copy()
groups  = day_pax.groupby('day_type')['airport_pax'].apply(list)
labels  = list(groups.index)
data    = [groups[lbl] for lbl in labels]

fig, ax = plt.subplots(figsize=(11, 5))
bp = ax.boxplot(data, labels=labels, patch_artist=True, notch=False,
                medianprops={'color': 'black', 'lw': 2})
palette = ['#3a86ff', '#ff9f1c', '#e63946', '#8ecae6', '#a8dadc']
for patch, clr in zip(bp['boxes'], palette[:len(labels)]):
    patch.set_facecolor(clr); patch.set_alpha(0.8)

ax.set_title('Distribucion de airport_pax por tipo de dia')
ax.set_xlabel('Tipo de dia'); ax.set_ylabel('Pasajeros aeropuerto')
ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'{x/1e6:.2f}M'))
plt.tight_layout(); plt.show()

print(day_pax.groupby('day_type')['airport_pax'].describe().round(0).to_string())


pivot_df = df.drop_duplicates('fecha')[['fecha','airport_pax','month','weekday']].copy()
heat     = pivot_df.pivot_table(values='airport_pax', index='month', columns='weekday', aggfunc='mean')
heat.columns = ['Lun','Mar','Mie','Jue','Vie','Sab','Dom']

cmap = LinearSegmentedColormap.from_list('blgr', ['#f7fbff', '#2171b5'])
fig, ax = plt.subplots(figsize=(11, 7))
im = ax.imshow(heat.values, cmap=cmap, aspect='auto')

ax.set_xticks(range(7)); ax.set_xticklabels(['Lun','Mar','Mie','Jue','Vie','Sab','Dom'])
ax.set_yticks(range(len(heat))); ax.set_yticklabels([f'Mes {m:02d}' for m in heat.index])

for i in range(heat.shape[0]):
    for j in range(heat.shape[1]):
        v = heat.values[i, j]
        if not np.isnan(v):
            ax.text(j, i, f'{v/1e6:.2f}M', ha='center', va='center', fontsize=7,
                    color='white' if v > heat.values[~np.isnan(heat.values)].max() * 0.6 else 'black')

plt.colorbar(im, ax=ax, label='Pax aeropuerto (media)')
ax.set_title('Heatmap: Pasajeros Aeropuerto -- Mes x Dia de Semana')
ax.set_xlabel('Dia de semana'); ax.set_ylabel('Mes')
plt.tight_layout(); plt.show()


ev_df = df.drop_duplicates('fecha')[['fecha','day_type','is_holiday','has_event','event_count',
                                     'max_impact_factor','airport_pax']].copy()

print("Dias con evento (has_event):")
print(ev_df['has_event'].value_counts().to_string())
print(f"\nmax_impact_factor -- min:{ev_df['max_impact_factor'].min():.3f}"
      f"  max:{ev_df['max_impact_factor'].max():.3f}"
      f"  mean:{ev_df['max_impact_factor'].mean():.3f}")

fig, axes = plt.subplots(1, 2, figsize=(13, 4))
for val, clr, lbl, ax in zip([False, True], ['#3a86ff', '#e63946'],
                               ['Sin evento', 'Con evento'], axes):
    sub = ev_df[ev_df['has_event'] == val]['airport_pax']
    ax.hist(sub, bins=30, color=clr, edgecolor='white', alpha=0.85)
    ax.axvline(sub.mean(), ls='--', color='black', lw=1.5,
               label=f'Media: {sub.mean()/1e6:.2f}M')
    ax.set_title(lbl); ax.set_xlabel('Airport pax'); ax.legend()
    ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'{int(x)}'))

plt.suptitle('Distribucion airport_pax: dias con y sin eventos', fontsize=13)
plt.tight_layout(); plt.show()

print("\nPax medio con/sin evento:")
print(ev_df.groupby('has_event')['airport_pax'].describe().round(0).to_string())


print("Zonas unicas:")
print(sorted(df['from_zone'].unique()))
print(f"\nSegmentos unicos: {df['segment_id'].nunique()}")
print(f"Paradas unicas (from_stop): {df['from_stop'].nunique()}")

zone_seg = df.groupby('from_zone')['segment_id'].nunique().sort_values(ascending=False)
print("\nSegmentos por zona:")
print(zone_seg.to_string())


zone_weight = df.groupby('from_zone')['from_node_weight'].mean().sort_values(ascending=False)
print("Node weight medio por zona:")
print(zone_weight.round(2).to_string())

fig, axes = plt.subplots(1, 2, figsize=(14, 4))
axes[0].barh(zone_seg.index, zone_seg.values, color='#3a86ff', edgecolor='white')
axes[0].set_title('Segmentos unicos por zona'); axes[0].set_xlabel('N segmentos')

axes[1].barh(zone_weight.index, zone_weight.values, color='#ff9f1c', edgecolor='white')
axes[1].set_title('Node weight medio por zona'); axes[1].set_xlabel('Node weight')

plt.tight_layout(); plt.show()


stops_geo = df.drop_duplicates('from_stop')[['from_stop','from_stop_name','from_lat','from_lon',
                                              'from_zone','from_node_weight','from_is_key_node']].copy()

palette10 = ['#3a86ff','#ff9f1c','#e63946','#8ecae6','#a8dadc',
             '#457b9d','#f4a261','#2dc653','#6a4c93','#b5838d']
zone_colors = {z: c for z, c in zip(stops_geo['from_zone'].unique(), palette10)}

fig, ax = plt.subplots(figsize=(11, 7))
for zone in stops_geo['from_zone'].unique():
    sub = stops_geo[stops_geo['from_zone'] == zone]
    ax.scatter(sub['from_lon'], sub['from_lat'],
               s=sub['from_node_weight'] * 20,
               c=zone_colors.get(zone, 'gray'),
               alpha=0.85, edgecolors='white', lw=0.7, label=zone)

# Annotate key nodes
for _, row in stops_geo[stops_geo['from_is_key_node']].iterrows():
    ax.annotate(row['from_stop_name'][:20], (row['from_lon'], row['from_lat']),
                fontsize=6.5, xytext=(5, 3), textcoords='offset points',
                arrowprops=dict(arrowstyle='-', color='gray', lw=0.5))

ax.set_title('Distribucion geografica de paradas -- Linea 1')
ax.set_xlabel('Longitud'); ax.set_ylabel('Latitud')
ax.legend(title='Zona', fontsize=8, loc='lower left')
plt.tight_layout(); plt.show()

print(f"\nParadas nodo clave (key_node): {stops_geo['from_is_key_node'].sum()}")


dir_counts = df.groupby('direction')['segment_id'].nunique()
print("Segmentos unicos por direccion:")
print(dir_counts.to_string())

seg_dir = df.groupby(['direction','segment_order'])['segment_distance_km'].mean()
print("\nDistancia media (km) por orden de segmento y direccion:")
print(seg_dir.round(3).to_string())

fig, ax = plt.subplots(figsize=(13, 4))
for direc, clr in zip(df['direction'].unique(), ['#3a86ff', '#e63946']):
    sub = df[df['direction'] == direc].groupby('segment_order')['segment_distance_km'].mean()
    ax.plot(sub.index, sub.values, 'o-', color=clr, lw=2, ms=6, label=direc)

ax.set_title('Distancia por segmento (km) -- Ida vs Vuelta')
ax.set_xlabel('Orden de segmento'); ax.set_ylabel('Km medio')
ax.legend(); plt.tight_layout(); plt.show()


# Relacion airport_pax en segmentos de zona Aeropuerto
airport_segs = df[df['from_zone'] == 'Aeropuerto'].drop_duplicates(['fecha','trip_id'])[['fecha','trip_id','trip_index','airport_pax']]
print(f"Registros de expediciones en zona Aeropuerto: {len(airport_segs)}")
print(f"\nairdport_pax por trip_index (media):")
print(airport_segs.groupby('trip_index')['airport_pax'].mean().describe().round(0).to_string())


day_feats = df.drop_duplicates('fecha').copy()
day_feats['is_holiday_n'] = day_feats['is_holiday'].astype(int)
day_feats['is_weekend_n'] = day_feats['is_weekend'].astype(int)
day_feats['has_event_n']  = day_feats['has_event'].astype(int)
day_feats['day_type_n']   = pd.Categorical(day_feats['day_type']).codes

corr_cols = ['airport_pax','temp_media','viento_kmh','lluvia_mm',
             'has_event_n','is_holiday_n','day_type_n',
             'pop_centro','pop_hondura','max_impact_factor','event_count']

corr_df = day_feats[corr_cols].dropna()
corr_m  = corr_df.corr()

cmap_div = LinearSegmentedColormap.from_list('rw_bl', ['#d62728','#ffffff','#1f77b4'])
fig, ax = plt.subplots(figsize=(12, 10))
im = ax.imshow(corr_m.values, cmap=cmap_div, vmin=-1, vmax=1, aspect='auto')

ax.set_xticks(range(len(corr_cols)))
ax.set_xticklabels(corr_cols, rotation=45, ha='right', fontsize=9)
ax.set_yticks(range(len(corr_cols)))
ax.set_yticklabels(corr_cols, fontsize=9)

for i in range(len(corr_cols)):
    for j in range(len(corr_cols)):
        v = corr_m.values[i, j]
        ax.text(j, i, f'{v:.2f}', ha='center', va='center', fontsize=7.5,
                color='white' if abs(v) > 0.6 else 'black')

plt.colorbar(im, ax=ax, label='Correlacion de Pearson')
ax.set_title('Matriz de correlacion -- variables para modelo D(t)')
plt.tight_layout(); plt.show()

target_corr = corr_m['airport_pax'].drop('airport_pax').sort_values(key=abs, ascending=False)
print("\nRanking de correlacion con airport_pax:")
print(target_corr.round(4).to_string())


try:
    from statsmodels.stats.outliers_influence import variance_inflation_factor
    from statsmodels.tools.tools import add_constant

    vif_input = corr_df.drop(columns=['airport_pax']).dropna()
    X = add_constant(vif_input)
    vif_data = pd.DataFrame({
        'Variable': X.columns,
        'VIF': [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]
    }).sort_values('VIF', ascending=False).query('Variable != "const"')

    print("Variance Inflation Factor (VIF):")
    print(vif_data.round(2).to_string(index=False))
    print("\nNota: VIF > 10 indica alta multicolinealidad | VIF > 5 moderada")

    fig, ax = plt.subplots(figsize=(10, 4))
    colors_vif = ['#e63946' if v > 10 else '#ff9f1c' if v > 5 else '#3a86ff'
                  for v in vif_data['VIF']]
    bars = ax.barh(vif_data['Variable'], vif_data['VIF'], color=colors_vif)
    ax.axvline(10, ls='--', color='crimson', lw=1.5, label='VIF=10 (alto)')
    ax.axvline(5,  ls='--', color='orange',  lw=1.5, label='VIF=5 (moderado)')
    ax.bar_label(bars, fmt='%.1f', padding=3, fontsize=9)
    ax.set_title('Variance Inflation Factor por variable'); ax.legend()
    plt.tight_layout(); plt.show()

except ImportError:
    print("statsmodels no esta disponible.")
    print("Instalar con: pip install statsmodels")
    print("Analisis VIF omitido.")


scatter_vars = ['temp_media', 'viento_kmh', 'lluvia_mm', 'max_impact_factor']
fig, axes = plt.subplots(2, 2, figsize=(14, 9))
axes = axes.flatten()

for i, var in enumerate(scatter_vars):
    sub = day_feats[['airport_pax', var]].dropna()
    axes[i].scatter(sub[var], sub['airport_pax'], alpha=0.35, s=18,
                    color='#3a86ff', edgecolors='none')
    m, b = np.polyfit(sub[var], sub['airport_pax'], 1)
    xs = np.linspace(sub[var].min(), sub[var].max(), 100)
    r  = sub.corr().iloc[0, 1]
    axes[i].plot(xs, m * xs + b, 'r-', lw=1.8, label=f'r={r:.3f}')
    axes[i].set_xlabel(var); axes[i].set_ylabel('airport_pax')
    axes[i].set_title(f'airport_pax vs {var}')
    axes[i].yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'{x/1e6:.1f}M'))
    axes[i].legend(fontsize=9)

plt.tight_layout(); plt.show()


meteo_vars = ['temp_media', 'viento_kmh', 'lluvia_mm']
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
colors_m = ['#e63946', '#3a86ff', '#00b4d8']

for ax, var, clr in zip(axes, meteo_vars, colors_m):
    data = day_feats[var].dropna()
    ax.hist(data, bins=35, color=clr, edgecolor='white', alpha=0.85)
    ax.axvline(data.mean(), ls='--', color='black', lw=1.5,
               label=f'media={data.mean():.1f}')
    ax.set_title(f'Distribucion {var}'); ax.set_xlabel(var); ax.legend()

plt.suptitle('Distribucion de variables meteorologicas', fontsize=13)
plt.tight_layout(); plt.show()

print(day_feats[meteo_vars].describe().round(2).to_string())


# Metricas operativas por dia con separacion explicita de unidades
ops_df = df.copy()
ops_df["cycle_id"] = ops_df["trip_id"].astype(str)
ops_df["is_service_segment"] = ops_df["direction"].isin(["ida", "vuelta"])
ops_df["service_id"] = np.where(
    ops_df["is_service_segment"],
    ops_df["cycle_id"] + "_" + ops_df["direction"].astype(str),
    np.nan,
)

sim = ops_df.groupby("fecha").agg(
    cycles_per_day      = ("cycle_id",                 "nunique"),
    services_per_day    = ("service_id",               "nunique"),
    km_per_day          = ("segment_distance_km",      "sum"),
    drive_time_total    = ("drive_time_min",           "sum"),
    dwell_time_total    = ("dwell_time_next_stop_min", "sum"),
    interval_used       = ("interval_min_used",        "first"),
    trips_report        = ("trips_total_per_day_report","first"),
    headway_report      = ("headway_min_report",       "first"),
    cycle_time_report   = ("cycle_time_min_report",    "first"),
    airport_pax         = ("airport_pax",              "first"),
).reset_index()

sim["total_time_total"] = sim["drive_time_total"] + sim["dwell_time_total"]
sim["services_per_cycle"] = sim["services_per_day"] / sim["cycles_per_day"]

print("=" * 65)
print("RESUMEN METRICAS OPERATIVAS DIARIAS")
print("=" * 65)
print(sim[["cycles_per_day","services_per_day","km_per_day","total_time_total","interval_used"]].describe().round(2).to_string())


print("Consistencia de unidad operativa (cycle_id -> service_id)")
print(sim["services_per_cycle"].value_counts().sort_index().rename("N dias").to_frame().to_string())

print("\nCiclos/dia (cycle_id):")
print(sim["cycles_per_day"].value_counts().sort_index().rename("N dias").to_frame().to_string())

print("\nExpediciones unidireccionales/dia (service_id):")
print(sim["services_per_day"].value_counts().sort_index().rename("N dias").to_frame().to_string())

print(f"\nIntervalo medio usado (dataset): {sim['interval_used'].mean():.3f} min")
print(f"Headway reportado (moda): {sim['headway_report'].mode()[0]} min")
print(f"Cycle time reportado (moda): {sim['cycle_time_report'].mode()[0]} min")

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, col, clr, lbl in [
    (axes[0], "services_per_day", "#3a86ff", "Expediciones unidireccionales/dia"),
    (axes[1], "km_per_day",       "#ff9f1c", "Km operados/dia"),
    (axes[2], "total_time_total", "#e63946", "Minutos operados/dia"),
]:
    ax.hist(sim[col], bins=30, color=clr, edgecolor="white", alpha=0.85)
    ax.axvline(sim[col].mean(), ls="--", color="black", lw=1.5,
               label=f"media={sim[col].mean():.1f}")
    ax.set_title(lbl); ax.set_xlabel(col); ax.legend(fontsize=9)

plt.suptitle("Distribucion de metricas operativas diarias", fontsize=13)
plt.tight_layout(); plt.show()


# Inspeccionar un ciclo representativo y sus expediciones
one_day = df["fecha"].min()
sample_cycle = df[df["fecha"] == one_day]["cycle_id"].iloc[0] if "cycle_id" in df.columns else df[df["fecha"] == one_day]["trip_id"].iloc[0]

sample_cycle_df = (
    df[(df["fecha"] == one_day) & (df["trip_id"] == sample_cycle)]
    .sort_values("segment_order")
    .copy()
)

print(f"Ciclo de muestra: {sample_cycle} | Fecha: {one_day.date()}")
cols_show = [
    "segment_order", "from_stop_name", "to_stop_name", "direction",
    "segment_distance_km", "drive_time_min", "dwell_time_next_stop_min"
]
pd.set_option("display.max_colwidth", 40)
print(sample_cycle_df[cols_show].to_string(index=False))

print(f"\nDistancia total ciclo      : {sample_cycle_df['segment_distance_km'].sum():.3f} km")
print(f"Tiempo conduccion ciclo    : {sample_cycle_df['drive_time_min'].sum():.2f} min")
print(f"Tiempo en paradas ciclo    : {sample_cycle_df['dwell_time_next_stop_min'].sum():.2f} min")
print(f"Tiempo total ciclo         : {(sample_cycle_df['drive_time_min'] + sample_cycle_df['dwell_time_next_stop_min']).sum():.2f} min")

for d in ["ida", "vuelta"]:
    leg = sample_cycle_df[sample_cycle_df["direction"] == d]
    print(f"\nExpedicion {d.upper()} del ciclo:")
    print(f"  Distancia: {leg['segment_distance_km'].sum():.3f} km")
    print(f"  Tiempo   : {(leg['drive_time_min'] + leg['dwell_time_next_stop_min']).sum():.2f} min")


fig, ax = plt.subplots(figsize=(11, 4))
iv = sim["interval_used"].dropna()
ax.hist(iv, bins=30, color="#6a4c93", edgecolor="white", alpha=0.85)
ax.axvline(iv.mean(), ls="--", color="crimson", lw=2, label=f"media={iv.mean():.3f} min")
ax.set_title("Distribucion de interval_min_used (inicio de ciclo)")
ax.set_xlabel("Minutos entre inicios de ciclo")
ax.legend()
plt.tight_layout(); plt.show()

print("\nDescripcion interval_min_used:")
print(iv.describe().round(4).to_string())


pax_s = df.drop_duplicates('fecha')['airport_pax'].dropna().sort_values()
Q1, Q3    = pax_s.quantile([0.25, 0.75])
IQR       = Q3 - Q1
lim_low   = Q1 - 3 * IQR
lim_high  = Q3 + 3 * IQR

day_check = df.drop_duplicates('fecha')[['fecha','day_type','airport_pax','has_event','is_holiday']].copy()
outlier_days = day_check[(day_check['airport_pax'] < lim_low) | (day_check['airport_pax'] > lim_high)]

print(f"Umbral outlier (3xIQR): < {lim_low:,.0f}  o  > {lim_high:,.0f}")
print(f"Dias con valores extremos: {len(outlier_days)}")
if len(outlier_days) > 0:
    print(outlier_days.to_string())

fig, ax = plt.subplots(figsize=(13, 4))
pax_series = df.drop_duplicates('fecha').set_index('fecha')['airport_pax'].sort_index()
ax.plot(pax_series.index, pax_series.values, color='#3a86ff', lw=0.8, alpha=0.7, label='airport_pax diario')
if len(outlier_days) > 0:
    ax.scatter(outlier_days['fecha'], outlier_days['airport_pax'],
               color='crimson', s=55, zorder=5, label='Outliers extremos (3xIQR)')
ax.axhline(lim_high, ls='--', color='orange', lw=1.2, label=f'Limite sup 3xIQR ({lim_high:,.0f})')
ax.axhline(lim_low,  ls='--', color='orange', lw=1.2, label=f'Limite inf 3xIQR ({lim_low:,.0f})')
ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'{x/1e6:.2f}M'))
ax.set_title('Serie airport_pax diario con deteccion de outliers')
ax.legend(fontsize=8); plt.tight_layout(); plt.show()


meteo_check = df.drop_duplicates('fecha')[['fecha','temp_media','viento_kmh','lluvia_mm','alerta']].copy()
null_dates  = meteo_check[meteo_check[['temp_media','viento_kmh','lluvia_mm']].isnull().any(axis=1)]

print(f"Dias con datos meteorologicos faltantes: {len(null_dates)}")
if len(null_dates) > 0:
    print(null_dates.to_string())
else:
    print("[OK] No hay valores faltantes en las variables meteorologicas")

print("\nDistribucion de alertas meteorologicas:")
print(df.drop_duplicates('fecha')['alerta'].value_counts().to_string())


neg_drive = (df['drive_time_min'] <= 0).sum()
neg_dwell = (df['dwell_time_next_stop_min'] < 0).sum()
print(f"Segmentos con drive_time_min <= 0 : {neg_drive}")
print(f"Segmentos con dwell_time_min < 0  : {neg_dwell}")

t_issues = df[(df['drive_time_min'] <= 0) | (df['dwell_time_next_stop_min'] < 0)]
if len(t_issues) > 0:
    print("\nMuestra de inconsistencias:")
    print(t_issues[['fecha','trip_id','segment_order','direction',
                     'drive_time_min','dwell_time_next_stop_min']].head(10).to_string())
else:
    print("[OK] No se encontraron tiempos negativos o nulos")

drive_q99 = df['drive_time_min'].quantile(0.99)
dwell_q99 = df['dwell_time_next_stop_min'].quantile(0.99)
print(f"\nSegmentos con drive_time > P99 ({drive_q99:.1f} min): {(df['drive_time_min'] > drive_q99).sum()}")
print(f"Segmentos con dwell_time > P99 ({dwell_q99:.1f} min): {(df['dwell_time_next_stop_min'] > dwell_q99).sum()}")

fig, axes = plt.subplots(1, 2, figsize=(14, 4))
axes[0].hist(df['drive_time_min'], bins=50, color='#3a86ff', edgecolor='white')
axes[0].axvline(drive_q99, color='crimson', ls='--', lw=1.5, label=f'P99={drive_q99:.1f}')
axes[0].set_title('Distribucion drive_time_min'); axes[0].set_xlabel('min'); axes[0].legend()

axes[1].hist(df['dwell_time_next_stop_min'], bins=50, color='#ff9f1c', edgecolor='white')
axes[1].axvline(dwell_q99, color='crimson', ls='--', lw=1.5, label=f'P99={dwell_q99:.1f}')
axes[1].set_title('Distribucion dwell_time_next_stop_min'); axes[1].set_xlabel('min'); axes[1].legend()

plt.tight_layout(); plt.show()


dir_check   = df.groupby('trip_id')['direction'].nunique()
mixed_trips = dir_check[dir_check > 1]
print(f"Viajes con mezcla de direcciones en un mismo trip_id: {len(mixed_trips)}")
if len(mixed_trips) > 0:
    print(df[df['trip_id'].isin(mixed_trips.index)].groupby('trip_id')['direction'].value_counts().head(20).to_string())
else:
    print("[OK] Cada trip_id tiene una unica direccion")

# Balance ida/vuelta por dia
dir_balance = df.groupby(['fecha','direction'])['segment_id'].count().unstack(fill_value=0)
if 'ida' in dir_balance.columns and 'vuelta' in dir_balance.columns:
    ratio = dir_balance['ida'] / dir_balance['vuelta']
    print("\nRatio ida/vuelta por dia (describe):")
    print(ratio.describe().round(3).to_string())


warnings.filterwarnings("ignore")

# PARAMETROS OPERATIVOS Y DE DEMANDA
PARAMS = {
    # -- Operacion --
    "CAPACITY": 55,
    "CYCLES_PER_DAY": 32,
    "SERVICES_PER_CYCLE": 2,
    "SERVICES_PER_DAY": 64,
    "CYCLE_MIN": 58,
    "KM_PER_CYCLE": 23.7,
    "KM_PER_SERVICE": 11.85,
    "HEADWAY_CYCLE_NOM": 30,

    # -- Perfil temporal de demanda por hora de salida (hipotesis operativa) --
    "hour_profile": {
        "06-07": 0.70,
        "07-09": 1.25,
        "09-13": 0.95,
        "13-16": 1.05,
        "16-20": 1.35,
        "20-23": 0.75,
    },

    # -- Tasas de adopcion alpha por escenario --
    "alpha_bajo":  0.010,
    "alpha_medio": 0.030,
    "alpha_alto":  0.050,

    # -- Factor beta por tipo de dia --
    "beta_day": {
        "laborable":       1.00,
        "sabado":          0.85,
        "domingo_festivo": 0.80,
    },

    # -- Peso relativo de paradas secundarias --
    "non_key_factor": 0.30,

    # -- Umbrales operativos --
    "sat_threshold":  1.00,
    "high_threshold": 0.80,
}

print("=" * 60)
print("PARAMETROS DEL MODELO (PARAMS)")
print("=" * 60)
for k, v in PARAMS.items():
    if k not in ["beta_day", "hour_profile"]:
        print(f"  {k:<25s}: {v}")

print("\n  beta_day (factor por tipo de dia):")
for dtype, val in PARAMS["beta_day"].items():
    print(f"    {dtype:<22s}: {val:.2f}")

print("\n  hour_profile (hipotesis de reparto intradiario):")
for k, v in PARAMS["hour_profile"].items():
    print(f"    {k:<8s}: {v:.2f}")


def compute_D(airport_pax, beta, gamma, alpha):
    """Fuente unica de verdad para D(t). Sin redondeo intermedio."""
    return float(alpha * airport_pax * beta * gamma)


# DataFrame a nivel dia (una fila unica por fecha)
day_df = (
    df.drop_duplicates(subset="fecha")
    .set_index("fecha")
    .sort_index()
    [["day_type", "airport_pax", "max_impact_factor", "has_event"]]
    .copy()
)

# beta_day: factor corrector por tipo de dia
day_df["beta"] = day_df["day_type"].map(PARAMS["beta_day"]).fillna(1.0)

# gamma_event: ya incorporado en max_impact_factor (1.0 si sin evento)
day_df["gamma"] = day_df["max_impact_factor"].fillna(1.0)

# D(t) para cada escenario (sin redondeo intermedio)
for scen, alpha in [("bajo", PARAMS["alpha_bajo"]),
                    ("medio", PARAMS["alpha_medio"]),
                    ("alto", PARAMS["alpha_alto"])]:
    day_df[f"D_{scen}"] = day_df.apply(
        lambda r: compute_D(r["airport_pax"], r["beta"], r["gamma"], alpha),
        axis=1
    )

print("=" * 65)
print("D(t) - PRIMERAS FILAS")
print("=" * 65)
cols_show = ["day_type", "airport_pax", "beta", "gamma", "D_bajo", "D_medio", "D_alto"]
print(day_df[cols_show].head(10).to_string())
print("\nEstadisticas de D(t) por escenario:")
print(day_df[["D_bajo", "D_medio", "D_alto"]].describe().round(1).to_string())
print("Nota: airport_pax es dato mensual. La variacion diaria proviene solo de beta_day y gamma_event.")


colors = {"bajo": "#3a86ff", "medio": "#ff9f1c", "alto": "#e63946"}
fig, axes = plt.subplots(2, 1, figsize=(15, 8), sharex=False)

ax = axes[0]
for scen, clr in colors.items():
    ax.plot(day_df.index, day_df[f"D_{scen}"], lw=0.9, alpha=0.85,
            color=clr, label=f"Escenario {scen}")
ax.set_title("Serie temporal de D(t) por escenario")
ax.set_ylabel("Pasajeros/dia (proxy)")
ax.legend(loc="upper left", framealpha=0.9)
ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f"{x:,.0f}"))

ax2 = axes[1]
day_types = sorted(day_df["day_type"].unique())
x_pos = np.arange(len(day_types))
width = 0.25
for i, (scen, clr) in enumerate(colors.items()):
    vals = [day_df[day_df["day_type"] == dt][f"D_{scen}"].values for dt in day_types]
    ax2.boxplot(vals, positions=x_pos + (i - 1) * width, widths=width * 0.85,
                patch_artist=True, notch=False,
                medianprops={"color": "black", "lw": 2},
                boxprops={"facecolor": clr, "alpha": 0.75},
                whiskerprops={"color": clr}, capprops={"color": clr},
                flierprops={"marker": ".", "color": clr, "alpha": 0.4})
ax2.set_xticks(x_pos)
ax2.set_xticklabels(day_types, rotation=15)
ax2.set_title("Distribucion de D(t) por tipo de dia")
ax2.set_ylabel("Pasajeros/dia (proxy)")
ax2.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f"{x:,.0f}"))
ax2.legend(handles=[Patch(facecolor=c, alpha=0.75, label=f"Escenario {s}")
                    for s, c in colors.items()], loc="upper right")
plt.suptitle("D(t) = alpha x airport_pax x beta_day x gamma_event", fontsize=13, y=1.01)
plt.tight_layout(); plt.show()

print("\nMedia D(t) por tipo de dia:")
print(day_df.groupby("day_type")[["D_bajo","D_medio","D_alto"]].mean().round(1).to_string())


stops = (
    df.drop_duplicates(subset="from_stop")
    [["from_stop", "from_stop_name", "from_seq", "from_zone",
      "from_is_key_node", "from_node_weight", "from_lat", "from_lon"]]
    .sort_values("from_seq").reset_index(drop=True).copy()
)

stops["w_raw"] = np.where(
    stops["from_is_key_node"],
    stops["from_node_weight"],
    stops["from_node_weight"] * PARAMS["non_key_factor"]
)
stops["w_norm"] = stops["w_raw"] / stops["w_raw"].sum()

print("=" * 70)
print("PESOS ESPACIALES NORMALIZADOS POR PARADA")
print("=" * 70)
print(stops[["from_stop_name","from_zone","from_is_key_node",
             "from_node_weight","w_raw","w_norm"]].to_string(index=False))
print(f"\nVerificacion: suma w_norm = {stops['w_norm'].sum():.6f}  (esperado: 1.0)")

fig, ax = plt.subplots(figsize=(14, 5))
colors_key = ["#e63946" if k else "#3a86ff" for k in stops["from_is_key_node"]]
bars = ax.bar(range(len(stops)), stops["w_norm"] * 100, color=colors_key, edgecolor="white")
ax.set_xticks(range(len(stops)))
ax.set_xticklabels(
    [f"{r['from_stop_name'][:18]}\n({r['from_zone'][:6]})" for _, r in stops.iterrows()],
    rotation=45, ha="right", fontsize=7.5)
ax.set_title("Peso espacial normalizado por parada (% de demanda asignada)")
ax.set_ylabel("w_i (%)")
ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f"{x:.1f}%"))
ax.legend(handles=[Patch(facecolor="#e63946", label="Key node"),
                   Patch(facecolor="#3a86ff", label="Parada secundaria")])
plt.tight_layout(); plt.show()


ZONES = ["Aeropuerto", "Puerto del Rosario", "La Hondura"]

OD_MATRIX = pd.DataFrame(
    [[0.00, 0.75, 0.25],
     [0.60, 0.00, 0.40],
     [0.30, 0.70, 0.00]],
    index=ZONES, columns=ZONES, dtype=float
)

assert (OD_MATRIX.sum(axis=1).round(6) == 1.0).all(), "ERROR: filas O-D no suman 1.0"

print("=" * 55)
print("MATRIZ O-D SINTETICA  P(destino | origen)")
print("=" * 55)
print(OD_MATRIX.to_string())
print("\nVerificacion - suma por fila:")
print(OD_MATRIX.sum(axis=1).to_string())

fig, ax = plt.subplots(figsize=(7, 5))
im = ax.imshow(OD_MATRIX.values, cmap="Blues", vmin=0, vmax=0.8)
ax.set_xticks(range(len(ZONES))); ax.set_xticklabels(["-> " + z for z in ZONES], rotation=20, ha="right")
ax.set_yticks(range(len(ZONES))); ax.set_yticklabels([z + " ->" for z in ZONES])
for i in range(len(ZONES)):
    for j in range(len(ZONES)):
        v = OD_MATRIX.values[i, j]
        ax.text(j, i, f"{v:.2f}", ha="center", va="center", fontsize=12,
                color="white" if v > 0.5 else "black", fontweight="bold")
plt.colorbar(im, ax=ax, label="P(destino | origen)")
ax.set_title("Matriz O-D sintetica - 3 zonas funcionales\nLinea 1 Puerto del Rosario")
plt.tight_layout(); plt.show()


# Preparar estructura base (unidad operativa: service_id)
seg_cols = [
    "fecha", "trip_id", "trip_index", "trip_start_time", "segment_id", "segment_order",
    "direction", "from_stop", "from_stop_name", "from_zone",
    "from_seq", "from_is_key_node", "from_node_weight",
    "to_stop", "to_stop_name", "to_zone", "drive_time_min", "dwell_time_next_stop_min"
]

base = (
    df[seg_cols]
    .drop_duplicates(subset=["fecha", "trip_id", "segment_order"])
    .sort_values(["fecha", "trip_id", "segment_order"])
    .reset_index(drop=True)
    .copy()
)

base["cycle_id"] = base["trip_id"].astype(str)
base["is_service_segment"] = base["direction"].isin(["ida", "vuelta"])

# Minutos desde medianoche del inicio de ciclo
t0 = pd.to_datetime(base["trip_start_time"], format="%H:%M:%S", errors="coerce")
base["cycle_start_min"] = t0.dt.hour * 60 + t0.dt.minute + t0.dt.second / 60
base["cycle_start_min"] = base["cycle_start_min"].fillna(0)

base["seg_total_min"] = base["drive_time_min"].fillna(0) + base["dwell_time_next_stop_min"].fillna(0)
base["cum_prev_min"] = (
    base.groupby(["fecha", "cycle_id"])["seg_total_min"].cumsum() - base["seg_total_min"]
)
base["seg_start_min"] = base["cycle_start_min"] + base["cum_prev_min"]

# Solo expediciones operativas (ida/vuelta). 'cierre' se excluye del modelo de carga.
sim_base = base[base["is_service_segment"]].copy()
sim_base["service_dir"] = sim_base["direction"].astype(str)
sim_base["service_id"] = sim_base["cycle_id"] + "_" + sim_base["service_dir"]
sim_base["service_seg_order"] = (
    sim_base.sort_values(["fecha", "service_id", "segment_order"])
    .groupby(["fecha", "service_id"]).cumcount() + 1
)

# Hora estimada de inicio de expedicion
service_start = (
    sim_base.groupby(["fecha", "service_id"], as_index=False)
    .agg(service_start_min=("seg_start_min", "min"),
         cycle_id=("cycle_id", "first"),
         service_dir=("service_dir", "first"),
         trip_index=("trip_index", "first"))
)
service_start = service_start.sort_values(["fecha", "service_start_min", "trip_index", "service_dir"])
service_start["service_seq_day"] = service_start.groupby("fecha").cumcount() + 1

sim_base = sim_base.merge(
    service_start[["fecha", "service_id", "service_start_min", "service_seq_day"]],
    on=["fecha", "service_id"], how="left"
)

w_map = stops.set_index("from_stop")["w_norm"].to_dict()
sim_base["w_norm"] = sim_base["from_stop"].map(w_map).fillna(0.001)

# Validaciones de unidad operativa
cycles_per_day = sim_base.groupby("fecha")["cycle_id"].nunique()
services_per_day = sim_base.groupby("fecha")["service_id"].nunique()
ratio = services_per_day / cycles_per_day
assert (ratio == PARAMS["SERVICES_PER_CYCLE"]).all(), "Inconsistencia cycle_id -> service_id"

print(f"Filas en sim_base                    : {len(sim_base):,}")
print(f"Fechas unicas                        : {sim_base['fecha'].nunique()}")
print(f"Ciclos por dia (median)              : {cycles_per_day.median():.0f}")
print(f"Expediciones unidireccionales por dia: {services_per_day.median():.0f}")
print(f"Segmentos/expedicion (media)         : {sim_base.groupby(['fecha','service_id'])['service_seg_order'].max().mean():.2f}")
print("[OK] Unidad operativa de simulacion: service_id (expedicion unidireccional).")


def od_fraction(origin_zone, dest_zone):
    """Fraccion P(dest_zone | origin_zone) segun la matriz O-D."""
    if origin_zone not in OD_MATRIX.index or dest_zone not in OD_MATRIX.columns:
        return 0.0
    return float(OD_MATRIX.loc[origin_zone, dest_zone])


def _parse_hour_profile(profile_dict):
    parsed = []
    for k, w in profile_dict.items():
        h0, h1 = k.split("-")
        parsed.append((int(h0), int(h1), float(w)))
    return parsed


def _hour_weight(minute_of_day, parsed_profile):
    h = int(minute_of_day // 60)
    for h0, h1, w in parsed_profile:
        if h0 <= h < h1:
            return w
    return 1.0


def _allocate_integers(total, weights):
    """Asignacion entera conservativa por restos mayores."""
    # Conserva exactamente el total tras redondear cuotas.
    weights = np.asarray(weights, dtype=float)
    if total <= 0 or len(weights) == 0:
        return np.zeros(len(weights), dtype=int)
    s = weights.sum()
    if s <= 0:
        weights = np.ones(len(weights), dtype=float)
        s = weights.sum()
    quotas = total * (weights / s)
    base = np.floor(quotas).astype(int)
    rem = int(total - base.sum())
    if rem > 0:
        idx = np.argsort(-(quotas - base))[:rem]
        base[idx] += 1
    return base


def build_service_demand(fecha, df_day, d_total_eq, hour_profile):
    """
    Reparte demanda diaria por expedicion con perfil temporal parametrizable.
    Politica: demanda equivalente diaria -> entero de pasajeros simulados.
    """
    d_total = int(round(float(d_total_eq)))

    services = (
        df_day.groupby(["service_id", "service_seq_day", "service_start_min"], as_index=False)
        .size()
        .sort_values(["service_seq_day", "service_start_min", "service_id"])
    )
    parsed_profile = _parse_hour_profile(hour_profile)
    services["w_time"] = services["service_start_min"].apply(lambda m: _hour_weight(m, parsed_profile))

    alloc = _allocate_integers(d_total, services["w_time"].values)
    services["d_service"] = alloc

    assert int(services["d_service"].sum()) == d_total, f"{fecha}: demanda diaria no conservada"

    return services[["service_id", "d_service"]], d_total


def simulate_day(fecha, df_day, d_total_eq, capacity, hour_profile):
    """
    Simula ocupacion por segmento para expediciones unidireccionales (service_id).
    """
    results = []
    checks = {
        "fecha": fecha,
        "d_total_int": 0,
        "d_assigned_services": 0,
        "board_desired_total": 0,
        "board_eff_total": 0,
        "alight_total": 0,
        "no_serv_total": 0,
        "past_alight_viol": 0,
    }

    svc_dem_df, d_total_int = build_service_demand(fecha, df_day, d_total_eq, hour_profile)
    checks["d_total_int"] = int(d_total_int)
    checks["d_assigned_services"] = int(svc_dem_df["d_service"].sum())

    for _, svc in svc_dem_df.iterrows():
        service_id = svc["service_id"]
        d_service = int(svc["d_service"])

        s_df = (
            df_day[df_day["service_id"] == service_id]
            .sort_values("service_seg_order")
            .copy()
        )
        if s_df.empty:
            continue

        # Ultimo segmento no admite boardings (sin futuro para bajar)
        boardable = s_df[s_df["service_seg_order"] < s_df["service_seg_order"].max()].copy()

        desired_by_seg = {int(so): 0 for so in s_df["service_seg_order"].tolist()}
        if d_service > 0 and len(boardable) > 0:
            alloc_stop = _allocate_integers(d_service, boardable["w_norm"].values)
            for so, n in zip(boardable["service_seg_order"].astype(int).tolist(), alloc_stop.tolist()):
                desired_by_seg[int(so)] = int(n)

        onboard_by_drop = {}
        occ = 0

        for _, row in s_df.iterrows():
            so = int(row["service_seg_order"])
            alight = int(onboard_by_drop.pop(so, 0))
            occ = occ - alight
            assert occ >= 0, f"{fecha} {service_id}: ocupacion negativa tras bajadas"

            board_desired = int(desired_by_seg.get(so, 0))
            checks["board_desired_total"] += board_desired

            available = max(int(capacity) - occ, 0)
            board_eff = min(board_desired, available)
            no_serv = board_desired - board_eff

            if board_eff > 0:
                future_df = s_df[s_df["service_seg_order"] > so]
                future_orders = future_df["service_seg_order"].astype(int).tolist()

                if len(future_orders) == 0:
                    no_serv += board_eff
                    board_eff = 0
                else:
                    fracs = np.array([od_fraction(row["from_zone"], z) for z in ZONES], dtype=float)
                    if fracs.sum() <= 0:
                        fracs = np.ones(len(ZONES), dtype=float)
                    alloc_zone = _allocate_integers(board_eff, fracs)

                    for z, n_z in zip(ZONES, alloc_zone.tolist()):
                        if n_z <= 0:
                            continue
                        cand = future_df[future_df["from_zone"] == z]["service_seg_order"].astype(int).tolist()
                        if len(cand) > 0:
                            chosen = min(cand)
                        else:
                            # Fallback seguro: forzar bajada al ultimo segmento futuro del mismo service_id.
                            chosen = max(future_orders)

                        if chosen <= so:
                            checks["past_alight_viol"] += 1
                            raise AssertionError(f"{fecha} {service_id}: alighting en pasado ({chosen} <= {so})")

                        onboard_by_drop[chosen] = onboard_by_drop.get(chosen, 0) + int(n_z)

            occ_potential = occ + board_desired
            # occ_potential incluye demanda no servida en el segmento; no equivale a ocupacion efectiva a bordo.
            occ = occ + board_eff

            checks["board_eff_total"] += int(board_eff)
            checks["no_serv_total"] += int(no_serv)
            checks["alight_total"] += int(alight)

            assert occ >= 0, f"{fecha} {service_id}: ocupacion negativa"
            assert no_serv >= 0, f"{fecha} {service_id}: demanda no servida negativa"

            results.append({
                "fecha": fecha,
                "cycle_id": row["cycle_id"],
                "service_id": service_id,
                "service_dir": row["service_dir"],
                "service_seq_day": int(row["service_seq_day"]),
                "service_seg_order": so,
                "from_stop": row["from_stop"],
                "from_stop_name": row["from_stop_name"],
                "from_zone": row["from_zone"],
                "boardings_deseados": int(board_desired),
                "boardings": int(board_eff),
                "alightings": int(alight),
                "occ": int(occ),
                "occ_potential": int(occ_potential),
                "saturado": bool(occ_potential > capacity),
                "no_servida": int(no_serv),
            })

        pax_remaining = int(sum(onboard_by_drop.values()))
        if pax_remaining != 0:
            raise AssertionError(f"{fecha} {service_id}: quedan {pax_remaining} pax a bordo al final")

    out = pd.DataFrame(results)

    if not out.empty:
        assert (out["occ"] >= 0).all(), f"{fecha}: occ negativa"
        assert (out["no_servida"] >= 0).all(), f"{fecha}: no_servida negativa"

    assert checks["d_assigned_services"] == checks["d_total_int"], f"{fecha}: reparto diario inconsistente"
    assert checks["board_desired_total"] == checks["board_eff_total"] + checks["no_serv_total"], f"{fecha}: boardings no conservados"
    assert checks["alight_total"] == checks["board_eff_total"], f"{fecha}: alightings no corresponden a boardings efectivos"
    assert checks["past_alight_viol"] == 0, f"{fecha}: alightings en el pasado detectados"

    return out, checks


def compute_kpis(sim_df, capacity):
    """KPIs separados por nivel: segmento vs expedicion."""
    if sim_df.empty:
        return {
            "segment": {
                "occ_mean_segment": 0.0,
                "occ_p95_segment": 0.0,
                "occ_max_segment": 0.0,
                "mean_segment_occupancy_ratio": 0.0,
                "seg_over80_pct": 0.0,
                "seg_saturated_pct": 0.0,
            },
            "expedition": {
                "exp_saturated_pct_potential": 0.0,
                "exp_max_occ_eff_mean": 0.0,
                "exp_max_occ_eff_p95": 0.0,
                "exp_max_occ_pot_p95": 0.0,
                "dem_no_servida_eq_period_total": 0,
                "dem_no_servida_eq_media_diaria": 0.0,
                "dem_no_servida_eq_share_pct": 0.0,
                "period_days": 0,
            },
        }

    # Se evita que la direccion con mas segmentos arrastre los promedios globales.
    dir_groups = [g for _, g in sim_df.groupby("service_dir")]
    if len(dir_groups) > 0:
        occ_mean = float(np.mean([g["occ"].mean() for g in dir_groups]))
        occ_p95 = float(np.mean([g["occ"].quantile(0.95) for g in dir_groups]))
        occ_max = float(np.max([g["occ"].max() for g in dir_groups]))
        seg_over80 = float(np.mean([(g["occ"] > capacity * PARAMS["high_threshold"]).mean() * 100 for g in dir_groups]))
        seg_sat = float(np.mean([(g["occ_potential"] > capacity).mean() * 100 for g in dir_groups]))
    else:
        occ_mean = float(sim_df["occ"].mean())
        occ_p95 = float(sim_df["occ"].quantile(0.95))
        occ_max = float(sim_df["occ"].max())
        seg_over80 = float((sim_df["occ"] > capacity * PARAMS["high_threshold"]).mean() * 100)
        seg_sat = float((sim_df["occ_potential"] > capacity).mean() * 100)

    mean_ratio = (occ_mean / float(capacity)) if capacity else np.nan

    # Ocupacion maxima por expedicion (efectiva) vs saturacion (potencial).
    exp_peak_eff = sim_df.groupby(["fecha", "service_id"])["occ"].max()
    exp_peak_pot = sim_df.groupby(["fecha", "service_id"])["occ_potential"].max()
    exp_sat_pot = float((exp_peak_pot > capacity).mean() * 100) if len(exp_peak_pot) else 0.0
    exp_peak_eff_mean = float(exp_peak_eff.mean()) if len(exp_peak_eff) else 0.0
    exp_peak_eff_p95 = float(exp_peak_eff.quantile(0.95)) if len(exp_peak_eff) else 0.0
    exp_peak_pot_p95 = float(exp_peak_pot.quantile(0.95)) if len(exp_peak_pot) else 0.0

    period_days = int(sim_df["fecha"].nunique())
    dem_no_serv_period = int(sim_df["no_servida"].sum())
    dem_no_serv_daily = float(dem_no_serv_period / period_days) if period_days > 0 else 0.0
    board_desired_total = int(sim_df["boardings_deseados"].sum())
    dem_no_serv_share = float(dem_no_serv_period / board_desired_total * 100) if board_desired_total > 0 else 0.0

    return {
        "segment": {
            "occ_mean_segment": round(occ_mean, 2),
            "occ_p95_segment": round(occ_p95, 2),
            "occ_max_segment": round(occ_max, 2),
            "mean_segment_occupancy_ratio": round(float(mean_ratio), 4),
            "seg_over80_pct": round(seg_over80, 2),
            "seg_saturated_pct": round(seg_sat, 2),
        },
        "expedition": {
            "exp_saturated_pct_potential": round(exp_sat_pot, 2),
            "exp_max_occ_eff_mean": round(exp_peak_eff_mean, 2),
            "exp_max_occ_eff_p95": round(exp_peak_eff_p95, 2),
            "exp_max_occ_pot_p95": round(exp_peak_pot_p95, 2),
            "dem_no_servida_eq_period_total": dem_no_serv_period,
            "dem_no_servida_eq_media_diaria": round(dem_no_serv_daily, 2),
            "dem_no_servida_eq_share_pct": round(dem_no_serv_share, 2),
            "period_days": period_days,
        },
    }

print("[OK] Funciones definidas: reparto temporal, simulacion por service_id y KPIs separados por nivel.")


sim_results = {}
validation_rows = []

for scen in ["bajo", "medio", "alto"]:
    print(f"\nSimulando escenario {scen.upper()}...")
    records = []
    fechas = sim_base["fecha"].drop_duplicates().sort_values().tolist()
    n = len(fechas)

    for idx, fecha in enumerate(fechas, start=1):
        if idx % 100 == 0 or idx == n:
            print(f"  {idx}/{n} fechas...", end="\r")

        df_day = sim_base[sim_base["fecha"] == fecha].copy()
        d_total = float(day_df.loc[fecha, f"D_{scen}"]) if fecha in day_df.index else 0.0

        day_sim, checks = simulate_day(
            fecha=fecha,
            df_day=df_day,
            d_total_eq=d_total,
            capacity=PARAMS["CAPACITY"],
            hour_profile=PARAMS["hour_profile"],
        )
        day_sim["escenario"] = scen
        records.append(day_sim)

        checks["escenario"] = scen
        validation_rows.append(checks)

    sim_results[scen] = pd.concat(records, ignore_index=True)
    print(f"\n[OK] {scen}: {len(sim_results[scen]):,} registros")

val_df = pd.DataFrame(validation_rows)
print("\n[VALIDACIONES INTERNAS]")
print(val_df.groupby("escenario")[["past_alight_viol","d_total_int","d_assigned_services","board_desired_total","board_eff_total","no_serv_total"]]
      .sum().to_string())
print("Nota: no_serv_total se expresa en demanda equivalente sintetica (proxy), no en pasajeros observados reales.")

assert (val_df["past_alight_viol"] == 0).all(), "Se detectaron alightings en el pasado"
assert (val_df["d_total_int"] == val_df["d_assigned_services"]).all(), "Demanda diaria mal repartida"
assert (val_df["board_desired_total"] == val_df["board_eff_total"] + val_df["no_serv_total"]).all(), "Conservacion de boardings incumplida"

print("\n[OK] Simulacion completa y validada para los tres escenarios.")


kpi_seg_rows = []
kpi_exp_rows = []
cap = PARAMS["CAPACITY"]

for scen, sdf in sim_results.items():
    kpis = compute_kpis(sdf, cap)

    seg_row = {"Escenario": scen}
    seg_row.update(kpis["segment"])
    kpi_seg_rows.append(seg_row)

    exp_row = {"Escenario": scen}
    exp_row.update(kpis["expedition"])
    kpi_exp_rows.append(exp_row)

kpi_seg_df = pd.DataFrame(kpi_seg_rows).set_index("Escenario")
kpi_exp_df = pd.DataFrame(kpi_exp_rows).set_index("Escenario")

print("=" * 92)
print("KPIs POR SEGMENTO - ESCENARIOS")
print("=" * 92)
print(kpi_seg_df.to_string())

print("\n" + "=" * 92)
print("KPIs POR EXPEDICION - ESCENARIOS (ocupacion efectiva y saturacion potencial)")
print("=" * 92)
print(kpi_exp_df.to_string())
print("\nNota KPI: *_eff_* usa ocupacion efectiva (occ). *_potential/*_pot_* usa ocupacion potencial (occ_potential).")
print("Nota KPI: dem_no_servida_eq_* esta en demanda equivalente sintetica, no en pasajeros observados reales.")

print("\n" + "=" * 78)
print("TOP 10 SEGMENTOS MAS CRITICOS (Escenario MEDIO)")
print("=" * 78)
top_segs = (
    sim_results["medio"]
    .groupby(["from_stop_name", "from_zone", "service_dir"])
    .agg(occ_media=("occ", "mean"),
         occ_p95=("occ", lambda x: x.quantile(0.95)),
         n_sat_seg=("saturado", "sum"),
         dem_no_servida_eq=("no_servida", "sum"))
    .round(2)
    .sort_values(["occ_media", "occ_p95"], ascending=False)
    .head(10)
    .reset_index()
)
print(top_segs.to_string(index=False))


cap = PARAMS["CAPACITY"]
fig, axes = plt.subplots(1, 3, figsize=(16, 5))
escenarios = ["bajo", "medio", "alto"]
palette = ["#3a86ff", "#ff9f1c", "#e63946"]

ax = axes[0]
occ_medias = [kpi_seg_df.loc[s, "occ_mean_segment"] for s in escenarios]
occ_p95s = [kpi_seg_df.loc[s, "occ_p95_segment"] for s in escenarios]
x = np.arange(len(escenarios)); w = 0.35
bars1 = ax.bar(x - w/2, occ_medias, w, color=palette, alpha=0.85, label="Media segmento")
bars2 = ax.bar(x + w/2, occ_p95s, w, color=palette, alpha=0.4, hatch="///", label="P95 segmento")
ax.axhline(cap, ls="--", color="crimson", lw=2, label=f"Capacidad ({cap})")
ax.axhline(cap * PARAMS["high_threshold"], ls=":", color="orange", lw=1.5,
           label=f"80% cap ({cap*PARAMS['high_threshold']:.0f})")
ax.bar_label(bars1, fmt="%.1f", padding=3, fontsize=9)
ax.bar_label(bars2, fmt="%.1f", padding=3, fontsize=9)
ax.set_xticks(x); ax.set_xticklabels([s.capitalize() for s in escenarios])
ax.set_title("Ocupacion por segmento (media y P95)")
ax.set_ylabel("Pasajeros")
ax.legend(fontsize=8)

ax = axes[1]
pct_sats = [kpi_exp_df.loc[s, "exp_saturated_pct_potential"] for s in escenarios]
bars3 = ax.bar(escenarios, pct_sats, color=palette, alpha=0.85, edgecolor="white")
ax.axhline(5, ls="--", color="crimson", lw=2, label="Umbral 5%")
ax.bar_label(bars3, fmt="%.2f%%", padding=3, fontsize=10)
ax.set_title("% Expediciones saturadas (segun occ_potential)")
ax.set_ylabel("% de expediciones")
ax.legend(fontsize=9)

ax = axes[2]
no_serv = [kpi_exp_df.loc[s, "dem_no_servida_eq_period_total"] for s in escenarios]
bars4 = ax.bar(escenarios, no_serv, color=palette, alpha=0.85, edgecolor="white")
ax.bar_label(bars4, fmt="%.0f", padding=3, fontsize=10)
ax.set_title("Demanda equivalente no servida acumulada (periodo)")
ax.set_ylabel("Unidades equivalentes (proxy)")

plt.suptitle("KPIs Operativos - Segmento y Expedicion (sin mezclar niveles)", fontsize=13)
plt.tight_layout(); plt.show()


print("\n=== SANITY CHECK: estructura de resultados por escenario ===")
for scen in ["bajo", "medio", "alto"]:
    sdf = sim_results[scen]
    cnts = sdf["service_dir"].value_counts(dropna=False)
    print(f"  {scen:>5s}: dirs={cnts.to_dict()} | services={sdf.groupby(['fecha','service_id']).ngroups}")

# Validaciones globales de integridad
for scen in ["bajo", "medio", "alto"]:
    sdf = sim_results[scen]
    assert (sdf["occ"] >= 0).all(), f"{scen}: occ negativa"
    assert (sdf["no_servida"] >= 0).all(), f"{scen}: no_servida negativa"

    peak = sdf.groupby(["fecha", "service_id"])["occ_potential"].max()
    assert len(peak) > 0, f"{scen}: sin expediciones para KPI"

    total_board = sdf["boardings"].sum()
    total_no_serv = sdf["no_servida"].sum()
    print(f"  {scen:>5s} | boardings efectivos={total_board:,} | dem_no_servida_eq={total_no_serv:,}")

fig, axes = plt.subplots(1, 2, figsize=(16, 6))
cap = PARAMS["CAPACITY"]

for di, dir_filter in enumerate(["ida", "vuelta"]):
    ax = axes[di]
    for scen, clr in zip(["bajo", "medio", "alto"], ["#3a86ff", "#ff9f1c", "#e63946"]):
        df_dir = sim_results[scen][sim_results[scen]["service_dir"] == dir_filter]
        if df_dir.empty:
            ax.text(0.5, 0.5, f"Sin datos {dir_filter}", transform=ax.transAxes, ha="center")
            continue
        prof = df_dir.groupby("service_seg_order")["occ"].mean().sort_index()
        ax.plot(prof.index, prof.values, "o-", color=clr, lw=2, ms=5,
                label=f"Escenario {scen} (mean={prof.mean():.2f})")

    ax.axhline(cap, ls="--", color="crimson", lw=2, label=f"Capacidad ({cap})")
    ax.axhline(cap * 0.80, ls=":", color="orange", lw=1.5, label="80% cap")
    ax.set_title(f"Perfil de ocupacion media - {dir_filter.upper()}")
    ax.set_xlabel("service_seg_order")
    ax.set_ylabel("Pasajeros")
    ax.legend(fontsize=8)

plt.suptitle("Perfil espacial de ocupacion por sentido y escenario", fontsize=13)
plt.tight_layout(); plt.show()


fig, axes = plt.subplots(1, 3, figsize=(16, 4))
cap = PARAMS["CAPACITY"]

for ax, scen, clr in zip(axes, ["bajo", "medio", "alto"], ["#3a86ff", "#ff9f1c", "#e63946"]):
    sdf = sim_results[scen]
    occ_pct = sdf["occ"] / cap * 100

    # KPI por segmento: porcentaje de segmentos potencialmente saturados
    seg_sat_pct = (sdf["occ_potential"] > cap).mean() * 100
    high_seg_pct = (sdf["occ"] > 0.8 * cap).mean() * 100

    # KPI por expedicion: porcentaje de expediciones saturadas
    exp_peak = sdf.groupby(["fecha", "service_id"])["occ_potential"].max()
    exp_sat_pct = (exp_peak > cap).mean() * 100 if len(exp_peak) else 0.0

    ax.hist(occ_pct, bins=40, color=clr, edgecolor="white", alpha=0.85)
    ax.axvline(100, ls="--", color="crimson", lw=2, label="100% (saturacion)")
    ax.axvline(80, ls=":", color="orange", lw=1.5, label="80% (presion alta)")
    ax.set_title(
        f"Escenario {scen.capitalize()}\n"
        f">80% segmentos: {high_seg_pct:.1f}% | % segmentos saturados: {seg_sat_pct:.1f}% | "
        f"% expediciones saturadas: {exp_sat_pct:.1f}%"
    )
    ax.set_xlabel("Ocupacion (% capacidad)")
    ax.set_ylabel("N segmentos")
    ax.legend(fontsize=8)

plt.suptitle("Distribucion de ocupacion como % de capacidad", fontsize=13)
plt.tight_layout(); plt.show()


# Sensibilidad operativa: variacion de capacidad/adopcion/frecuencia
base_services_day = int(sim_base.groupby("fecha")["service_id"].nunique().median())
alpha_medio = PARAMS["alpha_medio"]

sensitivity_variants = {
    f"Base ({base_services_day} exp/dia, alpha=0.030)": {
        "capacity": 55,
        "alpha": alpha_medio,
        "headway_dir_min": None,
        "schedule_mode": "observado",
    },
    "Capacidad 70 pax": {
        "capacity": 70,
        "alpha": alpha_medio,
        "headway_dir_min": None,
        "schedule_mode": "observado",
    },
    "Adopcion -10% (alpha=0.027)": {
        "capacity": 55,
        "alpha": alpha_medio * 0.9,
        "headway_dir_min": None,
        "schedule_mode": "observado",
    },
    "Adopcion +10% (alpha=0.033)": {
        "capacity": 55,
        "alpha": alpha_medio * 1.1,
        "headway_dir_min": None,
        "schedule_mode": "observado",
    },
    "Frecuencia ajustada (30 min/dir)": {
        "capacity": 55,
        "alpha": alpha_medio,
        "headway_dir_min": 30,
        "schedule_mode": "generado_por_headway",
    },
}


def compute_headway_by_direction(df_day):
    """Headway real por direccion calculado desde horarios de salida efectivos."""
    out = {}
    for d in sorted(df_day["service_dir"].dropna().unique().tolist()):
        starts = (
            df_day[df_day["service_dir"] == d]
            .groupby("service_id")["service_start_min"]
            .min()
            .sort_values()
            .values
        )
        out[d] = float(np.median(np.diff(starts))) if len(starts) > 1 else np.nan
        # Mediana para amortiguar huecos extremos de inicio/fin de servicio.
    return out


def build_frequency_schedule(df_day, headway_dir_min):
    if headway_dir_min is None or headway_dir_min <= 0:
        return df_day.copy()

    fecha_v = pd.to_datetime(df_day["fecha"].iloc[0])
    out = []

    for d in sorted(df_day["service_dir"].dropna().unique().tolist()):
        dir_df = df_day[df_day["service_dir"] == d].copy()
        svc_start = (
            dir_df.groupby("service_id", as_index=False)["service_start_min"]
            .min()
            .sort_values("service_start_min")
        )
        if svc_start.empty:
            continue

        win_start = float(svc_start["service_start_min"].min())
        win_end = float(svc_start["service_start_min"].max())

        starts = np.arange(win_start, win_end + 1e-9, float(headway_dir_min))
        if len(starts) == 0:
            starts = np.array([win_start])

        template_sid = svc_start.iloc[len(svc_start) // 2]["service_id"]
        template = dir_df[dir_df["service_id"] == template_sid].sort_values("service_seg_order").copy()
        template_start = float(template["service_start_min"].min())

        for j, st in enumerate(starts, start=1):
            t = template.copy()
            new_sid = f"{fecha_v.strftime('%Y%m%d')}_FRQ_{d}_{j:03d}"
            t["service_id"] = new_sid
            t["cycle_id"] = f"{fecha_v.strftime('%Y%m%d')}_FRQ_C_{d}_{j:03d}"
            t["service_start_min"] = float(st)
            if "seg_start_min" in t.columns:
                t["seg_start_min"] = t["seg_start_min"] + (float(st) - template_start)
            out.append(t)

    out_df = pd.concat(out, ignore_index=True)
    svc_order = (
        out_df.groupby("service_id", as_index=False)
        .agg(service_start_min=("service_start_min", "min"))
        .sort_values("service_start_min")
        .reset_index(drop=True)
    )
    svc_order["service_seq_day"] = np.arange(1, len(svc_order) + 1)

    out_df = (
        out_df.drop(columns=["service_seq_day"], errors="ignore")
        .merge(svc_order[["service_id", "service_seq_day"]], on="service_id", how="left")
    )
    return out_df


sens_rows = []
for variant_name, pv in sensitivity_variants.items():
    cap_v = pv["capacity"]
    alpha_v = pv["alpha"]
    headway_target = pv["headway_dir_min"]

    d_v = day_df.apply(
        lambda r: compute_D(r["airport_pax"], r["beta"], r["gamma"], alpha_v),
        axis=1
    )

    records_v = []
    checks_v = []
    hw_ida_list = []
    hw_vuelta_list = []
    exp_day_list = []

    for fecha in sim_base["fecha"].drop_duplicates().sort_values().tolist():
        df_day_base = sim_base[sim_base["fecha"] == fecha].copy()
        df_day_s = df_day_base if headway_target is None else build_frequency_schedule(df_day_base, headway_target)

        hw_map = compute_headway_by_direction(df_day_s)
        hw_ida_list.append(hw_map.get("ida", np.nan))
        hw_vuelta_list.append(hw_map.get("vuelta", np.nan))
        exp_day_list.append(int(df_day_s["service_id"].nunique()))

        d_t = float(d_v[fecha]) if fecha in d_v.index else 0.0
        day_sim_v, chk = simulate_day(
            fecha=fecha,
            df_day=df_day_s,
            d_total_eq=d_t,
            capacity=cap_v,
            hour_profile=PARAMS["hour_profile"],
        )
        records_v.append(day_sim_v)
        checks_v.append(chk)

    df_v = pd.concat(records_v, ignore_index=True)
    chk_v_df = pd.DataFrame(checks_v)
    assert (chk_v_df["d_total_int"] == chk_v_df["d_assigned_services"]).all(), f"{variant_name}: reparto demanda inconsistente"

    kpis_v = compute_kpis(df_v, cap_v)

    sens_rows.append({
        "Variante": variant_name,
        "schedule_mode": pv["schedule_mode"],
        "capacity": cap_v,
        "expeditions_per_day": round(float(np.median(exp_day_list)), 1),
        "headway_ida_min_real": round(float(np.nanmedian(hw_ida_list)), 2) if np.isfinite(np.nanmedian(hw_ida_list)) else np.nan,
        "headway_vuelta_min_real": round(float(np.nanmedian(hw_vuelta_list)), 2) if np.isfinite(np.nanmedian(hw_vuelta_list)) else np.nan,
        "occ_mean_segment": kpis_v["segment"]["occ_mean_segment"],
        "occ_p95_segment": kpis_v["segment"]["occ_p95_segment"],
        "mean_segment_occupancy_ratio": kpis_v["segment"]["mean_segment_occupancy_ratio"],
        "exp_saturated_pct_potential": kpis_v["expedition"]["exp_saturated_pct_potential"],
        "exp_max_occ_eff_p95": kpis_v["expedition"]["exp_max_occ_eff_p95"],
        "exp_max_occ_pot_p95": kpis_v["expedition"]["exp_max_occ_pot_p95"],
        "dem_no_servida_eq_period_total": kpis_v["expedition"]["dem_no_servida_eq_period_total"],
        "dem_no_servida_eq_media_diaria": kpis_v["expedition"]["dem_no_servida_eq_media_diaria"],
        "dem_no_servida_eq_share_pct": kpis_v["expedition"]["dem_no_servida_eq_share_pct"],
        "period_days": kpis_v["expedition"]["period_days"],
    })
    print(f"  [OK] {variant_name}")

sens_df = pd.DataFrame(sens_rows).set_index("Variante")

sens_schedule_df = sens_df[[
    "schedule_mode", "capacity", "expeditions_per_day", "headway_ida_min_real", "headway_vuelta_min_real"
]]

sens_segment_df = sens_df[[
    "occ_mean_segment", "occ_p95_segment", "mean_segment_occupancy_ratio"
]]

sens_expedition_df = sens_df[[
    "exp_saturated_pct_potential", "exp_max_occ_eff_p95", "exp_max_occ_pot_p95",
    "dem_no_servida_eq_period_total", "dem_no_servida_eq_media_diaria", "dem_no_servida_eq_share_pct", "period_days"
]]

print("\n" + "=" * 104)
print("ANALISIS DE SENSIBILIDAD - PARAMETROS DE HORARIO (por direccion)")
print("=" * 104)
print(sens_schedule_df.to_string())

print("\n" + "=" * 104)
print("ANALISIS DE SENSIBILIDAD - KPIs POR SEGMENTO")
print("=" * 104)
print(sens_segment_df.to_string())

print("\n" + "=" * 104)
print("ANALISIS DE SENSIBILIDAD - KPIs POR EXPEDICION")
print("=" * 104)
print(sens_expedition_df.to_string())
print("\nNota KPI sensibilidad: *_eff_* usa ocupacion efectiva, *_potential/*_pot_* usa ocupacion potencial.")
print("Nota KPI sensibilidad: dem_no_servida_eq_* es demanda equivalente sintetica (proxy), no pasajeros observados reales.")


fig, axes = plt.subplots(1, 3, figsize=(16, 5))
variants = list(sens_df.index)
palette_s = ["#264653", "#2a9d8f", "#e9c46a", "#f4a261", "#e76f51"]

ax = axes[0]
bars = ax.barh(variants, sens_df["occ_mean_segment"].values, color=palette_s, edgecolor="white", alpha=0.85)
ax.axvline(PARAMS["CAPACITY"], ls="--", color="crimson", lw=2, label=f"Cap. base {PARAMS['CAPACITY']}")
ax.axvline(70, ls=":", color="steelblue", lw=1.5, label="Cap. 70")
ax.bar_label(bars, fmt="%.2f", padding=3, fontsize=9)
ax.set_title("Ocupacion media por segmento")
ax.legend(fontsize=8)

ax = axes[1]
bars2 = ax.barh(variants, sens_df["exp_saturated_pct_potential"].values, color=palette_s, edgecolor="white", alpha=0.85)
ax.axvline(5, ls="--", color="crimson", lw=2, label="Umbral 5%")
ax.bar_label(bars2, fmt="%.2f%%", padding=3, fontsize=9)
ax.set_title("% Expediciones saturadas (segun occ_potential)")
ax.legend(fontsize=9)

ax = axes[2]
bars3 = ax.barh(variants, sens_df["dem_no_servida_eq_period_total"].values, color=palette_s, edgecolor="white", alpha=0.85)
ax.bar_label(bars3, fmt="%.0f", padding=3, fontsize=9)
ax.set_title("Demanda equivalente no servida acumulada (periodo)")

plt.suptitle("Analisis de sensibilidad - KPIs segmentarios y por expedicion", fontsize=12)
plt.tight_layout(); plt.show()


# Factores diarios y dataset a nivel fecha
# D_index se definira como la normalizacion del mismo proxy base que
# alimenta la simulacion: airport_pax * beta_day * gamma_event.
BETA_DAY = {
    "laborable":       1.00,
    "sabado":          0.85,
    "domingo_festivo": 0.80,
}

daily_cols = [
    "fecha", "year", "month", "day", "weekday", "weekday_name",
    "day_type", "is_holiday", "is_weekend", "has_event",
    "airport_pax", "max_impact_factor", "event_count",
    "temp_media", "viento_kmh", "lluvia_mm", "alerta",
    "hay_clases", "periodo", "pop_centro", "pop_hondura",
]
daily_cols = [c for c in daily_cols if c in df.columns]

df_daily = (
    df.sort_values("fecha")
    .groupby("fecha")[daily_cols]
    .first()
    .reset_index(drop=True)
    .copy()
)

df_daily["fecha"] = pd.to_datetime(df_daily["fecha"])
df_daily = df_daily.sort_values("fecha").reset_index(drop=True)

df_daily["beta"] = df_daily["day_type"].map(BETA_DAY).fillna(1.0)
df_daily["gamma"] = df_daily["max_impact_factor"].fillna(1.0)

print(f"df_daily: {df_daily.shape[0]} filas  |  {df_daily['fecha'].min().date()} -> {df_daily['fecha'].max().date()}")
print(f"Anios disponibles: {sorted(df_daily['fecha'].dt.year.unique())}")
print(f"day_type disponibles: {sorted(df_daily['day_type'].dropna().unique())}")


# CONSTRUCCION DE D_index alineado con la simulacion operativa
# proxy_raw = airport_pax * beta_day * gamma_event
# D_index   = normalizacion MinMax del proxy_raw sobre train (2024)

df_daily["proxy_raw"] = (
    df_daily["airport_pax"] * df_daily["beta"] * df_daily["gamma"]
)

train_mask = df_daily["fecha"].dt.year == 2024
test_mask  = df_daily["fecha"].dt.year == 2025

df_train = df_daily[train_mask].copy()
df_test  = df_daily[test_mask].copy()

print(f"Train (2024): {len(df_train)} dias")
print(f"Test  (2025): {len(df_test)} dias")

proxy_min = df_train["proxy_raw"].min()
proxy_max = df_train["proxy_raw"].max()
proxy_rng = proxy_max - proxy_min if proxy_max != proxy_min else 1.0

df_daily["D_index"] = ((df_daily["proxy_raw"] - proxy_min) / proxy_rng).clip(0, 1)

df_train = df_daily[train_mask].copy()
df_test  = df_daily[test_mask].copy()

print(f"\nproxy_raw [train]: min={proxy_min:,.1f}  max={proxy_max:,.1f}")
print(f"D_index [train] : min={df_train['D_index'].min():.4f}  max={df_train['D_index'].max():.4f}  mean={df_train['D_index'].mean():.4f}")
print(f"D_index [test]  : min={df_test['D_index'].min():.4f}  max={df_test['D_index'].max():.4f}  mean={df_test['D_index'].mean():.4f}")


# Visualizacion 1: Serie temporal D_index (train / test)
fig, ax = plt.subplots(figsize=(15, 4))

ax.plot(df_train["fecha"], df_train["D_index"], color="#3a86ff", lw=0.9, alpha=0.85,
        label="Train (2024)")
ax.plot(df_test["fecha"],  df_test["D_index"],  color="#e63946", lw=0.9, alpha=0.85,
        label="Test (2025)")

# Linea de corte
cutoff = df_test["fecha"].min()
ax.axvline(cutoff, ls="--", color="black", lw=1.5, label=f"Corte train/test ({cutoff.date()})")

# Media movil 30d
rolling = df_daily.set_index("fecha")["D_index"].rolling(30, center=True).mean()
ax.plot(rolling.index, rolling.values, color="dimgray", lw=1.5, ls=":", label="Media movil 30d")

ax.set_title("Serie temporal de D_index (train / test)")
ax.set_ylabel("D_index (0-1)"); ax.legend(fontsize=9)
plt.tight_layout(); plt.show()

# Tabla: estadisticos de D_index por tipo de dia
print("\nD_index por tipo de dia (conjunto completo):")
tbl = (df_daily.groupby("day_type")["D_index"]
       .agg(["count","mean","std","min","max"])
       .round(4)
       .rename(columns={"count":"N","mean":"Media","std":"Std","min":"Min","max":"Max"}))
print(tbl.to_string())



# Reproducibilidad global
SEED_GLOBAL = 16
MODEL_RANDOM_STATE = 17
# Semillas globales para reproducibilidad de experimentos y comparativas.
os.environ["PYTHONHASHSEED"] = str(SEED_GLOBAL)
np.random.seed(SEED_GLOBAL)
random.seed(SEED_GLOBAL)

MODEL_DIR = Path("model_artifacts")
MODEL_DIR.mkdir(parents=True, exist_ok=True)

nan_pct_hc = df_daily["hay_clases"].isna().mean() * 100
print(f"hay_clases NaN: {nan_pct_hc:.1f}%")

if nan_pct_hc > 30:
    df_daily["hay_clases_bin"] = df_daily["hay_clases"].apply(lambda x: 0 if (pd.isna(x) or x == 0) else 1)
    hc_feature = "hay_clases_bin"
    print("  -> Binarizada como hay_clases_bin (0/1, NaN tratado como 0)")
else:
    df_daily["hay_clases_bin"] = df_daily["hay_clases"].fillna(0)
    hc_feature = "hay_clases_bin"
    print("  -> Imputada con 0 (NaN < 30%)")

NUM_FEATURES = [
    "airport_pax",
    "temp_media", "viento_kmh", "lluvia_mm",
    "month", "weekday", hc_feature,
]
CAT_FEATURES = ["day_type"]
# max_impact_factor excluido: correlacion perfecta con has_event (r=1.0).
# Se mantiene has_event como senal binaria de evento.
BIN_FEATURES = ["is_holiday", "has_event"]
ALL_FEATURES = NUM_FEATURES + CAT_FEATURES + BIN_FEATURES

# Target para entrenamiento CV: proxy_raw (evita fuga por normalizacion fold)
TARGET_RAW = "proxy_raw"
TARGET_IDX = "D_index"

train_mask = df_daily["fecha"].dt.year == 2024
test_mask  = df_daily["fecha"].dt.year == 2025

X_train = df_daily.loc[train_mask, ALL_FEATURES].copy()
X_test  = df_daily.loc[test_mask,  ALL_FEATURES].copy()

y_train_raw = df_daily.loc[train_mask, TARGET_RAW].copy()
y_test_raw  = df_daily.loc[test_mask,  TARGET_RAW].copy()
y_train_idx = df_daily.loc[train_mask, TARGET_IDX].copy()
y_test_idx  = df_daily.loc[test_mask,  TARGET_IDX].copy()

for b in BIN_FEATURES:
    X_train[b] = X_train[b].fillna(0).astype(int)
    X_test[b]  = X_test[b].fillna(0).astype(int)


def raw_to_dindex(y_raw):
    # Clip obligatorio para asegurar rango fisico [0,1] en evaluacion e inferencia.
    y = np.asarray(y_raw, dtype=float)
    return np.clip((y - proxy_min) / proxy_rng, 0.0, 1.0)


def cv_mae_dindex_from_raw(estimator, X, y_raw, splitter):
    """CV MAE en escala D_index aplicando la misma transformacion (incluye clipping)."""
    # Se evalua en la misma escala final del reporte para evitar comparaciones CV/Test inconsistentes.
    fold_mae = []
    for tr_idx, va_idx in splitter.split(X):
        est = clone(estimator)
        X_tr, X_va = X.iloc[tr_idx], X.iloc[va_idx]
        y_tr, y_va = y_raw.iloc[tr_idx], y_raw.iloc[va_idx]
        est.fit(X_tr, y_tr)
        pred_va_raw = est.predict(X_va)
        y_va_idx = raw_to_dindex(y_va.values)
        pred_va_idx = raw_to_dindex(pred_va_raw)
        fold_mae.append(mean_absolute_error(y_va_idx, pred_va_idx))
    return float(np.mean(fold_mae)), float(np.std(fold_mae))


print(f"\nX_train: {X_train.shape}  |  y_train_raw: {y_train_raw.shape}")
print(f"X_test : {X_test.shape}   |  y_test_raw : {y_test_raw.shape}")
print(f"\nFeatures numericas : {NUM_FEATURES}")
print(f"Feature categorica : {CAT_FEATURES}")
print(f"Features binarias  : {BIN_FEATURES}")
print(f"\nSemillas fijadas -> global={SEED_GLOBAL} | model_random_state={MODEL_RANDOM_STATE}")
print(f"Cache de modelos: {MODEL_DIR.resolve()}")
print("ALCANCE ML: emulador del indice proxy sintetico (no predictor operativo de demanda real).")

# Chequeo de redundancia de variables de evento
if "has_event" in df_daily.columns:
    corr_evt = np.corrcoef(
        df_daily["has_event"].astype(float),
        df_daily["max_impact_factor"].astype(float)
    )[0, 1]
    print(f"\nCorrelacion has_event vs max_impact_factor: {corr_evt:.4f}")

    features_in_use = set(NUM_FEATURES + CAT_FEATURES + BIN_FEATURES)
    if ("has_event" in features_in_use) and ("max_impact_factor" in features_in_use):
        assert abs(corr_evt) < 0.9999, "Redundancia casi perfecta de variables de evento"
    else:
        print("  -> Info: alta correlacion detectada, pero no se usan ambas como features simultaneas.")


# Baseline: media de proxy_raw por day_type en TRAIN -> aplicar a TEST
mean_by_day_type_raw = y_train_raw.groupby(X_train["day_type"]).mean()
print("Media de proxy_raw por day_type en train:")
print(mean_by_day_type_raw.round(2).to_string())

y_baseline_raw = X_test["day_type"].map(mean_by_day_type_raw).fillna(y_train_raw.mean())
y_baseline = raw_to_dindex(y_baseline_raw)

mae_base = mean_absolute_error(y_test_idx, y_baseline)
rmse_base = np.sqrt(mean_squared_error(y_test_idx, y_baseline))

print("\nBaseline (proxy_raw por day_type -> D_index):")
print(f"  MAE  = {mae_base:.4f}")
print(f"  RMSE = {rmse_base:.4f}")



numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler",  StandardScaler()),
])

categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot",  OneHotEncoder(handle_unknown="ignore", sparse_output=False)),
])

binary_transformer = SimpleImputer(strategy="constant", fill_value=0)

preprocessor = ColumnTransformer([
    ("num", numeric_transformer,     NUM_FEATURES),
    ("cat", categorical_transformer, CAT_FEATURES),
    ("bin", binary_transformer,      BIN_FEATURES),
])

tscv = TimeSeriesSplit(n_splits=5)
alphas_grid = [0.01, 0.1, 1.0, 10.0, 100.0]

ridge_artifact_path = MODEL_DIR / "ridge_artifact_seed16_rs17_v1.joblib"

if ridge_artifact_path.exists():
    ridge_bundle = joblib.load(ridge_artifact_path)
    ridge_search = ridge_bundle["search"]
    cv_mae_ridge_dindex = ridge_bundle["cv_mae_dindex"]
    cv_std_ridge_dindex = ridge_bundle["cv_std_dindex"]
    print(f"[LOAD] Ridge cargado desde {ridge_artifact_path.name}")
else:
    ridge_search = GridSearchCV(
        Pipeline([
            ("prep",  preprocessor),
            ("ridge", Ridge()),
        ]),
        param_grid={"ridge__alpha": alphas_grid},
        cv=tscv,
        scoring="neg_mean_absolute_error",
        n_jobs=1,
        refit=True,
    )
    ridge_search.fit(X_train, y_train_raw)
    cv_mae_ridge_dindex, cv_std_ridge_dindex = cv_mae_dindex_from_raw(ridge_search.best_estimator_, X_train, y_train_raw, tscv)

    ridge_bundle = {
        "search": ridge_search,
        "cv_mae_dindex": cv_mae_ridge_dindex,
        "cv_std_dindex": cv_std_ridge_dindex,
    }
    joblib.dump(ridge_bundle, ridge_artifact_path)
    print(f"[SAVE] Ridge guardado en {ridge_artifact_path.name}")

ridge_pipeline = ridge_search.best_estimator_
best_alpha = ridge_search.best_params_["ridge__alpha"]
print(f"Alpha optimo (TimeSeriesSplit-5): {best_alpha}")
print(f"CV MAE Ridge (D_index): {cv_mae_ridge_dindex:.4f} Â± {cv_std_ridge_dindex:.4f}")

y_pred_ridge_raw = ridge_pipeline.predict(X_test)
y_pred_ridge = raw_to_dindex(y_pred_ridge_raw)

mae_ridge  = mean_absolute_error(y_test_idx, y_pred_ridge)
rmse_ridge = np.sqrt(mean_squared_error(y_test_idx, y_pred_ridge))
r2_ridge   = r2_score(y_test_idx, y_pred_ridge)

print("\n" + "=" * 55)
print("METRICAS EN TEST (2025) sobre D_index sintetico (emulacion de proxy)")
print("=" * 55)
print(f"{'Modelo':<20} {'MAE':>8}  {'RMSE':>8}  {'R2':>8}")
print("-" * 55)
print(f"{'Baseline (day_type)':<20} {mae_base:>8.4f}  {rmse_base:>8.4f}  {'--':>8}")
print(f"{'Ridge':<20} {mae_ridge:>8.4f}  {rmse_ridge:>8.4f}  {r2_ridge:>8.4f}")

mejora_mae  = (mae_base  - mae_ridge)  / mae_base  * 100
mejora_rmse = (rmse_base - rmse_ridge) / rmse_base * 100
print(f"\nMejora Ridge vs Baseline: MAE {mejora_mae:+.1f}%  |  RMSE {mejora_rmse:+.1f}%")


# Coeficientes mas influyentes
# Recuperar nombres de features tras ColumnTransformer
num_feat_names = NUM_FEATURES
cat_feat_names = list(
    ridge_pipeline.named_steps["prep"]
    .named_transformers_["cat"]
    .named_steps["onehot"]
    .get_feature_names_out(CAT_FEATURES)
)
bin_feat_names = BIN_FEATURES
all_feat_names = num_feat_names + cat_feat_names + bin_feat_names

coefs = ridge_pipeline.named_steps["ridge"].coef_
coef_df = (
    pd.DataFrame({"Feature": all_feat_names, "Coeficiente": coefs})
    .assign(AbsCoef=lambda d: d["Coeficiente"].abs())
    .sort_values("AbsCoef", ascending=False)
    .head(10)
    .reset_index(drop=True)
)

print("=" * 55)
print("TOP 10 COEFICIENTES (valor absoluto) - Ridge")
print("=" * 55)
print(coef_df[["Feature","Coeficiente"]].to_string(index=False))


fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Panel 1: y_true vs y_pred (D_index de reporte)
ax = axes[0]
ax.scatter(y_test_idx, y_pred_ridge, alpha=0.45, s=25, color="#3a86ff", edgecolors="none")
lims = [min(y_test_idx.min(), y_pred_ridge.min()) - 0.02,
        max(y_test_idx.max(), y_pred_ridge.max()) + 0.02]
ax.plot(lims, lims, "r--", lw=1.5, label="y = x (prediccion perfecta)")
ax.set_xlabel("D_index real (test 2025)")
ax.set_ylabel("D_index predicho (Ridge)")
ax.set_title(f"Prediccion vs Real - Ridge\nMAE={mae_ridge:.4f}  RMSE={rmse_ridge:.4f}  R2={r2_ridge:.4f}")
ax.legend(fontsize=9)

# Panel 2: coeficientes top 10
ax = axes[1]
colors_coef = ["#e63946" if v > 0 else "#3a86ff" for v in coef_df["Coeficiente"]]
bars = ax.barh(coef_df["Feature"][::-1], coef_df["Coeficiente"][::-1],
               color=colors_coef[::-1], edgecolor="white", alpha=0.85)
ax.axvline(0, color="black", lw=0.8)
ax.set_title("Top 10 coeficientes Ridge\n(rojo=positivo  azul=negativo)")
ax.set_xlabel("Coeficiente")

plt.suptitle("Modelo 1 - Ridge | Evaluacion sobre D_index (proxy)", fontsize=13)
plt.tight_layout(); plt.show()


residuals = y_test_idx.values - y_pred_ridge

fig, axes = plt.subplots(1, 2, figsize=(13, 4))

ax = axes[0]
ax.scatter(y_pred_ridge, residuals, alpha=0.4, s=22, color="#ff9f1c", edgecolors="none")
ax.axhline(0, ls="--", color="crimson", lw=1.5)
ax.set_xlabel("D_index predicho")
ax.set_ylabel("Residuo (real - pred)")
ax.set_title("Residuos vs valores predichos")

ax = axes[1]
ax.hist(residuals, bins=35, color="#ff9f1c", edgecolor="white", alpha=0.85)
ax.axvline(0, ls="--", color="crimson", lw=1.5)
ax.axvline(residuals.mean(), ls=":", color="black", lw=1.5,
           label=f"Media={residuals.mean():.4f}")
ax.set_xlabel("Residuo")
ax.set_ylabel("Frecuencia")
ax.set_title("Distribucion de residuos - Ridge")
ax.legend()

plt.suptitle("Analisis de residuos - Ridge (proxy D_index)", fontsize=12)
plt.tight_layout(); plt.show()

print(f"\nResiduo medio : {residuals.mean():.4f}")
print(f"Residuo std   : {residuals.std():.4f}")



xgb_artifact_path = MODEL_DIR / "xgb_artifact_seed16_rs17_v1.joblib"

tscv = TimeSeriesSplit(n_splits=5)

if xgb_artifact_path.exists():
    xgb_bundle = joblib.load(xgb_artifact_path)
    gs_xgb = xgb_bundle["search"]
    cv_mae_xgb_raw = xgb_bundle["cv_mae_raw"]
    cv_mae_xgb_dindex = xgb_bundle["cv_mae_dindex"]
    cv_std_xgb_dindex = xgb_bundle["cv_std_dindex"]
    print(f"[LOAD] XGBoost cargado desde {xgb_artifact_path.name}")
else:
    xgb_pipeline = Pipeline([
        ("prep", preprocessor),
        ("xgb",  XGBRegressor(
            objective="reg:squarederror",
            random_state=MODEL_RANDOM_STATE,
            verbosity=0,
            n_jobs=1,
        )),
    ])

    param_grid_xgb = {
        "xgb__n_estimators":  [100, 300],
        "xgb__max_depth":     [3, 5],
        "xgb__learning_rate": [0.05, 0.1],
        "xgb__subsample":     [0.8, 1.0],
    }

    gs_xgb = GridSearchCV(
        xgb_pipeline,
        param_grid_xgb,
        cv=tscv,
        scoring="neg_mean_absolute_error",
        n_jobs=1,
        refit=True,
        verbose=0,
    )
    gs_xgb.fit(X_train, y_train_raw)

    cv_mae_xgb_raw = abs(gs_xgb.best_score_)
    cv_mae_xgb_dindex, cv_std_xgb_dindex = cv_mae_dindex_from_raw(gs_xgb.best_estimator_, X_train, y_train_raw, tscv)

    xgb_bundle = {
        "search": gs_xgb,
        "cv_mae_raw": cv_mae_xgb_raw,
        "cv_mae_dindex": cv_mae_xgb_dindex,
        "cv_std_dindex": cv_std_xgb_dindex,
    }
    joblib.dump(xgb_bundle, xgb_artifact_path)
    print(f"[SAVE] XGBoost guardado en {xgb_artifact_path.name}")

print(f"Mejores hiperparametros XGBoost: {gs_xgb.best_params_}")
print(f"MAE CV (proxy_raw): {cv_mae_xgb_raw:.4f}")
print(f"MAE CV (D_index) : {cv_mae_xgb_dindex:.4f} Â± {cv_std_xgb_dindex:.4f}")

y_pred_xgb_raw = gs_xgb.predict(X_test)
y_pred_xgb = raw_to_dindex(y_pred_xgb_raw)


mae_xgb  = mean_absolute_error(y_test_idx, y_pred_xgb)
rmse_xgb = np.sqrt(mean_squared_error(y_test_idx, y_pred_xgb))
r2_xgb   = r2_score(y_test_idx, y_pred_xgb)

print("\n" + "=" * 62)
print("METRICAS EN TEST (2025) sobre D_index sintetico (emulacion de proxy)")
print("=" * 62)
print(f"{'Modelo':<22} {'MAE':>8}  {'RMSE':>8}  {'R2':>8}")
print("-" * 62)
print(f"{'Baseline (day_type)':<22} {mae_base:>8.4f}  {rmse_base:>8.4f}  {'--':>8}")
print(f"{'Ridge':<22} {mae_ridge:>8.4f}  {rmse_ridge:>8.4f}  {r2_ridge:>8.4f}")
print(f"{'XGBoost':<22} {mae_xgb:>8.4f}  {rmse_xgb:>8.4f}  {r2_xgb:>8.4f}")

print("\nComparabilidad CV/Test en misma escala (D_index):")
print(f"  XGBoost CV MAE (D_index): {cv_mae_xgb_dindex:.4f}")
print(f"  XGBoost Test MAE (D_index): {mae_xgb:.4f}")



# Feature importance (gain) del mejor modelo XGBoost
best_xgb = gs_xgb.best_estimator_.named_steps["xgb"]
prep_fit  = gs_xgb.best_estimator_.named_steps["prep"]

# Nombres reales de features tal como los ve el modelo
# (ColumnTransformer con OneHotEncoder para day_type -> multiples columnas)
try:
    xgb_feat_names = list(prep_fit.get_feature_names_out())
    # Limpiar prefijos 'num__', 'cat__day_type_', 'bin__'
    xgb_feat_names = [
        n.replace("num__", "").replace("cat__", "").replace("bin__", "")
        for n in xgb_feat_names
    ]
except Exception:
    # Fallback: construir manualmente como ultimo recurso
    cat_names = list(
        prep_fit.named_transformers_["cat"]
        .named_steps["onehot"]
        .get_feature_names_out(CAT_FEATURES)
    )
    xgb_feat_names = NUM_FEATURES + cat_names + BIN_FEATURES

importance_vals = best_xgb.feature_importances_
n_feats_model   = len(importance_vals)
n_names         = len(xgb_feat_names)

if n_feats_model != n_names:
    print(f"AVISO: names={n_names}, importances={n_feats_model} -> truncando")
    xgb_feat_names = xgb_feat_names[:n_feats_model]

imp_df = (
    __import__("pandas").DataFrame({
        "Feature":    xgb_feat_names,
        "Importance": importance_vals,
    })
    .sort_values("Importance", ascending=False)
    .head(10)
    .reset_index(drop=True)
)

print("=" * 50)
print("FEATURE IMPORTANCE - XGBoost (gain, top 10)")
print("=" * 50)
print(imp_df.to_string(index=False))

# Visualizacion
fig, ax = plt.subplots(figsize=(11, 5))
palette_imp = ["#e63946" if i == 0 else "#3a86ff" for i in range(len(imp_df))]
bars = ax.barh(imp_df["Feature"][::-1], imp_df["Importance"][::-1],
               color=palette_imp[::-1], edgecolor="white", alpha=0.85)
ax.bar_label(bars, fmt="%.4f", padding=3, fontsize=9)
ax.set_title("Feature Importance \u2014 XGBoost (gain)\nTop 10 variables")
ax.set_xlabel("Importancia (gain relativo)")
plt.tight_layout(); plt.show()


warnings.filterwarnings("ignore")

best_params = gs_xgb.best_params_
n_est_range = [50, 100, 150, 200, 250, 300]
tscv5 = TimeSeriesSplit(n_splits=5)

mae_val_list = []
mae_train_list = []

for n_est in n_est_range:
    pip_lc = Pipeline([
        ("prep", preprocessor),
        ("xgb",  XGBRegressor(
            n_estimators  = n_est,
            max_depth     = best_params.get("xgb__max_depth", 3),
            learning_rate = best_params.get("xgb__learning_rate", 0.1),
            subsample     = best_params.get("xgb__subsample", 0.8),
            objective     = "reg:squarederror",
            random_state  = MODEL_RANDOM_STATE,
            verbosity     = 0,
            n_jobs        = 1,
        )),
    ])

    cv_scores = cross_val_score(pip_lc, X_train, y_train_raw,
                                cv=tscv5, scoring="neg_mean_absolute_error")
    mae_val_list.append(-cv_scores.mean())

    pip_lc.fit(X_train, y_train_raw)
    mae_train_list.append(mean_absolute_error(y_train_raw, pip_lc.predict(X_train)))

fig, ax = plt.subplots(figsize=(11, 5))
ax.plot(n_est_range, mae_train_list, "o-", color="#3a86ff", lw=2, ms=6, label="MAE Train (proxy_raw)")
ax.plot(n_est_range, mae_val_list,   "o-", color="#ff9f1c", lw=2, ms=6, label="MAE Validacion CV (proxy_raw)")
ax.axvline(best_params.get("xgb__n_estimators", 100), ls="--", color="crimson",
           lw=1.5, label=f"Mejor n_estimators={best_params.get('xgb__n_estimators',100)}")
ax.set_xlabel("n_estimators")
ax.set_ylabel("MAE")
ax.set_title("Curvas de aprendizaje - XGBoost (escala proxy_raw)")
ax.legend(fontsize=9)
plt.tight_layout(); plt.show()

print("\nMAE por n_estimators (proxy_raw):")
for n, tr, va in zip(n_est_range, mae_train_list, mae_val_list):
    print(f"  n={n:>3d}  train={tr:.4f}  val_cv={va:.4f}  gap={va-tr:.4f}")

gap_ratio = (mae_val_list[-1] / mae_train_list[-1]
             if mae_train_list[-1] > 0 else float("inf"))
if gap_ratio > 5:
    print(f"ADVERTENCIA SOBREAJUSTE: ratio val/train MAE = {gap_ratio:.1f}x. "
          f"XGBoost memoriza train y no generaliza de forma robusta.")

# Residuos del modelo en test: positivos = infraestimacion, negativos = sobreestimacion.
residuals_xgb = y_test_idx.values - y_pred_xgb

# Dos vistas complementarias: ajuste real-vs-predicho y distribucion del error.
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Panel 1: dispersion prediccion vs valor real.
ax = axes[0]
ax.scatter(y_test_idx, y_pred_xgb, alpha=0.45, s=25, color="#3a86ff", edgecolors="none")
# Limites comunes para dibujar la diagonal de prediccion perfecta (y=x).
lims = [min(float(y_test_idx.min()), float(y_pred_xgb.min())) - 0.02,
        max(float(y_test_idx.max()), float(y_pred_xgb.max())) + 0.02]
ax.plot(lims, lims, "r--", lw=1.5, label="y = x (prediccion perfecta)")
ax.set_xlabel("D_index real (test 2025)")
ax.set_ylabel("D_index predicho (XGBoost)")
ax.set_title(f"Prediccion vs Real - XGBoost\nMAE={mae_xgb:.4f}  RMSE={rmse_xgb:.4f}  R2={r2_xgb:.4f}")
ax.legend(fontsize=9)

# Panel 2: histograma de residuos para detectar sesgo y variabilidad.
ax = axes[1]
ax.hist(residuals_xgb, bins=35, color="#ff9f1c", edgecolor="white", alpha=0.85)
ax.axvline(0, ls="--", color="crimson", lw=1.5)
# La media del residuo resume sesgo medio del modelo en test.
ax.axvline(residuals_xgb.mean(), ls=":", color="black", lw=1.5,
           label=f"Media={residuals_xgb.mean():.4f}")
ax.set_xlabel("Residuo (real - predicho)")
ax.set_ylabel("Frecuencia")
ax.set_title("Distribucion de residuos - XGBoost")
ax.legend()

plt.suptitle("Modelo 2 - XGBoost | Evaluacion sobre D_index", fontsize=13)
plt.tight_layout(); plt.show()

print(f"\nResiduo medio : {residuals_xgb.mean():.4f}")
print(f"Residuo std   : {residuals_xgb.std():.4f}")



mlp_artifact_path = MODEL_DIR / "mlp_artifact_seed16_rs17_v1.joblib"

tscv = TimeSeriesSplit(n_splits=5)

if mlp_artifact_path.exists():
    mlp_bundle = joblib.load(mlp_artifact_path)
    gs_mlp = mlp_bundle["search"]
    cv_mae_mlp_raw = mlp_bundle["cv_mae_raw"]
    cv_mae_mlp_dindex = mlp_bundle["cv_mae_dindex"]
    cv_std_mlp_dindex = mlp_bundle["cv_std_dindex"]
    mlp_converged = mlp_bundle["mlp_converged"]
    print(f"[LOAD] MLP cargado desde {mlp_artifact_path.name}")
else:
    mlp_pipeline = Pipeline([
        ("prep", preprocessor),
        ("mlp",  MLPRegressor(
            random_state=MODEL_RANDOM_STATE,
            max_iter=5000,
            early_stopping=False,
            tol=1e-4,
        )),
    ])

    param_grid_mlp = {
        "mlp__hidden_layer_sizes": [(32,), (64,), (64, 32)],
        "mlp__alpha":              [0.001, 0.01, 0.05],
        "mlp__learning_rate_init": [0.0005, 0.001, 0.003],
    }

    gs_mlp = GridSearchCV(
        mlp_pipeline,
        param_grid_mlp,
        cv=tscv,
        scoring="neg_mean_absolute_error",
        n_jobs=1,
        refit=True,
        verbose=0,
    )
    gs_mlp.fit(X_train, y_train_raw)

    best_mlp_tmp = gs_mlp.best_estimator_.named_steps["mlp"]
    mlp_converged = bool(best_mlp_tmp.n_iter_ < best_mlp_tmp.max_iter)

    cv_mae_mlp_raw = abs(gs_mlp.best_score_)
    cv_mae_mlp_dindex, cv_std_mlp_dindex = cv_mae_dindex_from_raw(gs_mlp.best_estimator_, X_train, y_train_raw, tscv)

    mlp_bundle = {
        "search": gs_mlp,
        "cv_mae_raw": cv_mae_mlp_raw,
        "cv_mae_dindex": cv_mae_mlp_dindex,
        "cv_std_dindex": cv_std_mlp_dindex,
        "mlp_converged": mlp_converged,
    }
    joblib.dump(mlp_bundle, mlp_artifact_path)
    print(f"[SAVE] MLP guardado en {mlp_artifact_path.name}")

best_mlp = gs_mlp.best_estimator_.named_steps["mlp"]
print(f"Mejores hiperparametros MLP: {gs_mlp.best_params_}")
print(f"MAE CV (proxy_raw): {cv_mae_mlp_raw:.4f}")
print(f"CV MAE en escala D_index: {cv_mae_mlp_dindex:.4f} Â± {cv_std_mlp_dindex:.4f}")
print("ADVERTENCIA METODO: early_stopping desactivado para mantener validacion temporal estricta en TimeSeriesSplit.")

y_pred_mlp_raw = gs_mlp.predict(X_test)
y_pred_mlp = raw_to_dindex(y_pred_mlp_raw)


mae_mlp  = mean_absolute_error(y_test_idx, y_pred_mlp)
rmse_mlp = np.sqrt(mean_squared_error(y_test_idx, y_pred_mlp))
r2_mlp   = r2_score(y_test_idx, y_pred_mlp)

if not mlp_converged:
    print("ADVERTENCIA: MLP no convergio en el ajuste final. Resultado potencialmente inestable.")
else:
    print(f"MLP convergio en {best_mlp.n_iter_} iteraciones.")

mlp_cv_test_ratio = (mae_mlp / cv_mae_mlp_dindex) if cv_mae_mlp_dindex > 0 else np.nan
mlp_stable = bool(mlp_converged and (0.67 <= mlp_cv_test_ratio <= 1.50))

print(f"CV MAE MLP en escala D_index:   {cv_mae_mlp_dindex:.4f}")
print(f"Test MAE MLP en escala D_index: {mae_mlp:.4f}")
if pd.notna(mlp_cv_test_ratio):
    print(f"Ratio Test/CV MLP: {mlp_cv_test_ratio:.2f}x")
if not mlp_stable:
    print("ADVERTENCIA: gap CV/Test anomalo o no convergencia; MLP se penaliza en la comparativa final.")

print("\n" + "=" * 65)
print("METRICAS EN TEST (2025) - TABLA COMPARATIVA (EMULADOR DE PROXY)")
print("=" * 65)
print(f"{'Modelo':<22} {'MAE':>8}  {'RMSE':>8}  {'R2':>8}")
print("-" * 65)
print(f"{'Baseline (day_type)':<22} {mae_base:>8.4f}  {rmse_base:>8.4f}  {'--':>8}")
print(f"{'Ridge':<22} {mae_ridge:>8.4f}  {rmse_ridge:>8.4f}  {r2_ridge:>8.4f}")
print(f"{'XGBoost':<22} {mae_xgb:>8.4f}  {rmse_xgb:>8.4f}  {r2_xgb:>8.4f}")
print(f"{'MLP':<22} {mae_mlp:>8.4f}  {rmse_mlp:>8.4f}  {r2_mlp:>8.4f}")



fig, ax = plt.subplots(figsize=(11, 4))
ax.plot(best_mlp.loss_curve_, color="#3a86ff", lw=2, label="Train loss")
ax.set_xlabel("Epocas")
ax.set_ylabel("Loss")
ax.set_title(f"Curva de entrenamiento - MLP\nIters hasta convergencia: {best_mlp.n_iter_}")
ax.legend(fontsize=9)
plt.tight_layout(); plt.show()

print(f"Epocas ejecutadas (n_iter_): {best_mlp.n_iter_}")
print(f"Arquitectura optima: {gs_mlp.best_params_['mlp__hidden_layer_sizes']}")


residuals_mlp = y_test_idx.values - y_pred_mlp

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

ax = axes[0]
ax.scatter(y_test_idx, y_pred_mlp, alpha=0.45, s=25, color="#8338ec", edgecolors="none")
lims = [min(float(y_test_idx.min()), float(y_pred_mlp.min())) - 0.02,
        max(float(y_test_idx.max()), float(y_pred_mlp.max())) + 0.02]
ax.plot(lims, lims, "r--", lw=1.5, label="y = x (prediccion perfecta)")
ax.set_xlabel("D_index real (test 2025)")
ax.set_ylabel("D_index predicho (MLP)")
ax.set_title(f"Prediccion vs Real - MLP\nMAE={mae_mlp:.4f}  RMSE={rmse_mlp:.4f}  R2={r2_mlp:.4f}")
ax.legend(fontsize=9)

ax = axes[1]
ax.hist(residuals_mlp, bins=35, color="#8338ec", edgecolor="white", alpha=0.85)
ax.axvline(0, ls="--", color="crimson", lw=1.5)
ax.axvline(residuals_mlp.mean(), ls=":", color="black", lw=1.5,
           label=f"Media={residuals_mlp.mean():.4f}")
ax.set_xlabel("Residuo (real - predicho)")
ax.set_ylabel("Frecuencia")
ax.set_title("Distribucion de residuos - MLP")
ax.legend()

plt.suptitle("Modelo 3 - MLP | Evaluacion sobre D_index", fontsize=13)
plt.tight_layout(); plt.show()

print(f"\nResiduo medio : {residuals_mlp.mean():.4f}")
print(f"Residuo std   : {residuals_mlp.std():.4f}")



modelos = ["Baseline", "Ridge", "XGBoost", "MLP"]
maes = [mae_base, mae_ridge, mae_xgb, mae_mlp]
rmses = [rmse_base, rmse_ridge, rmse_xgb, rmse_mlp]
r2s = ["--", f"{r2_ridge:.4f}", f"{r2_xgb:.4f}", f"{r2_mlp:.4f}"]

cv_mae_cols = [np.nan, cv_mae_ridge_dindex, cv_mae_xgb_dindex, cv_mae_mlp_dindex]

gap_ratio_xgb = (mae_val_list[-1] / mae_train_list[-1]) if ("mae_val_list" in globals() and "mae_train_list" in globals() and mae_train_list[-1] > 0) else np.nan
xgb_obs = "Descartado: sobreajuste severo" if (pd.notna(gap_ratio_xgb) and gap_ratio_xgb > 10) else "No recomendado"

if not mlp_converged:
    mlp_obs = "Descartado: no converge"
elif not mlp_stable:
    mlp_obs = "Descartado: gap CV/Test anomalo"
else:
    mlp_obs = "No recomendado (inferior a Ridge)"

observaciones = [
    "Referencia",
    "Recomendado",
    xgb_obs,
    mlp_obs,
]
colores = ["#adb5bd", "#3a86ff", "#e63946", "#8338ec"]

x = np.arange(len(modelos))
w = 0.6

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
for ax, vals, metric in zip(axes, [maes, rmses], ["MAE", "RMSE"]):
    bars = ax.bar(x, vals, w, color=colores, edgecolor="white", alpha=0.88)
    ax.axhline(vals[0], ls="--", color="#adb5bd", lw=1.5, label=f"Baseline {metric}={vals[0]:.4f}")
    ax.bar_label(bars, fmt="%.4f", padding=3, fontsize=10)
    ax.set_xticks(x); ax.set_xticklabels(modelos)
    ax.set_title(f"{metric} en test 2025 (D_index)")
    ax.set_ylabel(metric)
    ax.legend(fontsize=9)

plt.suptitle("Comparativa final de modelos (aproximacion de proxy)", fontsize=13)
plt.tight_layout(); plt.show()

print("\n" + "=" * 118)
print("COMPARATIVA DE MODELOS Ã¢â‚¬â€ ReconstrucciÃƒÂ³n de ÃƒÂ­ndice proxy sintÃƒÂ©tico")
print("=" * 118)
print(f"{'Modelo':<20} {'CV_MAE_Didx':>12}  {'Test_MAE_Didx':>13}  {'RMSE':>8}  {'R2':>8}  {'ObservaciÃƒÂ³n':<36}")
print("-" * 118)
for m, cvm, mae, rmse, r2, obs in zip(modelos, cv_mae_cols, maes, rmses, r2s, observaciones):
    cv_txt = "--" if pd.isna(cvm) else f"{cvm:.4f}"
    print(f"  {m:<18} {cv_txt:>12}  {mae:>13.4f}  {rmse:>8.4f}  {r2:>8}  {obs:<36}")

if pd.notna(gap_ratio_xgb):
    print(f"\nControl sobreajuste XGBoost: gap val/train = {gap_ratio_xgb:.1f}x")
if pd.notna(mlp_cv_test_ratio):
    print(f"Control estabilidad MLP: ratio Test/CV = {mlp_cv_test_ratio:.2f}x")

print("Nota: Este bloque ML funciona como emulador del indice proxy sintetico.")
print("Nota: RÃ‚Â² y MAE miden ajuste sobre D_index sintÃƒÂ©tico, no sobre demanda observada real.")



# Demo: emulacion para dias hipoteticos
# Salida: D_index clippeado y demanda equivalente por escenario (proxy).

dias_futuros = pd.DataFrame({
    "airport_pax":        [125000, 98000, 142000, 110000],
    "max_impact_factor":  [1.00,   1.00,  1.20,   1.00],
    "temp_media":         [22.0,   19.5,  24.0,   21.0],
    "viento_kmh":         [18.0,   35.0,  12.0,   22.0],
    "lluvia_mm":          [0.0,    2.5,   0.0,    0.0],
    "month":              [7,      11,    8,      3],
    "weekday":            [0,      5,     6,      2],
    "hay_clases_bin":     [1,      0,     0,      1],
    "day_type":           ["laborable", "sabado", "domingo_festivo", "laborable"],
    "is_holiday":         [0,      0,     0,      0],
    "has_event":          [0,      0,     1,      0],
})


if "has_event" not in dias_futuros.columns:
    dias_futuros["has_event"] = (dias_futuros["max_impact_factor"] > 1.0).astype(int)
dias_futuros["has_event"] = dias_futuros["has_event"].fillna(0).astype(int)

descripcion = [
    "Lunes verano (alta temporada)",
    "Sabado noviembre (viento fuerte)",
    "Domingo agosto con evento",
    "Miercoles marzo (temporada media)",
]

proxy_pred = ridge_pipeline.predict(dias_futuros)
d_index_pred = raw_to_dindex(proxy_pred)

# Validacion obligatoria: D_index en [0, 1]
assert ((d_index_pred >= 0) & (d_index_pred <= 1)).all(), "Predicciones D_index fuera de [0,1]"

proxy_raw_pred = d_index_pred * proxy_rng + proxy_min
proxy_raw_pred = np.maximum(proxy_raw_pred, 0)

alphas = {
    "bajo":  PARAMS["alpha_bajo"],
    "medio": PARAMS["alpha_medio"],
    "alto":  PARAMS["alpha_alto"],
}

print("=" * 96)
print("DEMO EMULADOR: D_index (clip) y demanda equivalente sintetica por escenario")
print("=" * 96)
print(f"{'Dia':<38} {'D_index':>8}  {'Proxy_raw':>10}  {'D_eq_bajo':>10}  {'D_eq_medio':>10}  {'D_eq_alto':>10}")
print("-" * 96)

for desc, d_idx, proxy in zip(descripcion, d_index_pred, proxy_raw_pred):
    d_b = proxy * alphas["bajo"]
    d_m = proxy * alphas["medio"]
    d_a = proxy * alphas["alto"]
    print(f"  {desc:<36} {d_idx:>8.3f}  {proxy:>10.1f}  {d_b:>10.1f}  {d_m:>10.1f}  {d_a:>10.1f}")

print("\n[OK] Todas las predicciones D_index quedaron acotadas en [0,1].")
print("Nota: resultados de emulacion del proxy sintetico; no son prediccion operativa de demanda real observada.")


# Verificacion rapida de dependencias del notebook

modules = {
    "pandas": "pandas",
    "numpy": "numpy",
    "matplotlib": "matplotlib",
    "sklearn": "sklearn",
    "scipy": "scipy",
    "seaborn": "seaborn",
    "statsmodels": "statsmodels",
    "xgboost": "xgboost",
    "jinja2": "jinja2",
}

print("DEPENDENCIAS - CHECK")
print("=" * 60)
for name, mod in modules.items():
    m = importlib.import_module(mod)
    ver = getattr(m, "__version__", "(sin __version__)")
    print(f"{name:12s} OK  version={ver}")

```
